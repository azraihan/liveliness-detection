```python
"""
gdfas_experiments.py
====================
GD-FAS ("Group-wise Scaling and Orthogonal Decomposition for Domain-Invariant
Feature Extraction in Face Anti-Spoofing", ICCV 2025 -- official repo, CLIP
ViT-B/16 backbone) run on the SAME three datasets, metrics, and experiment
protocol as the G2V2former notebook (lets-see-2.ipynb):

  EXPERIMENT 1: train on LCC-FASD only, test on all 3 datasets
  EXPERIMENT 2: train on LCC-FASD, then finetune on Celeb-DF-v2 (train split
                disjoint from the official test protocol), test on all 3
  Metrics: HTER / AUC per dataset; grouped bar charts + CSV comparison.

Run in the OFFLINE, no-internet GPU Kaggle notebook. Every cell is marked
`# %% CELL n` -- paste each into its own notebook cell, run top to bottom.

Inputs to attach (Kaggle "Add Data", no internet needed once attached):
  - gdfas_offline_bundle(.zip)  -- produced by prepare_gdfas_offline_bundle.py
    in a separate internet-enabled notebook: pip wheels (ftfy/regex), the
    CLIP ViT-B-16.pt checkpoint, and the GD-FAS repo itself.
  - LCC-FASD (faber24/lcc-fasd)
  - Celeb-DF-v2 (reubensuju/celeb-df-v2)
  - the Asian-Fakes stills (raihanzahin/prep-sbi-extra)
  - OPTIONAL: the published cache dataset from the G2V2former runs holding
    celebdf_index.pt / celebdf_train_index.pt -- reusing them keeps the
    exact same Celeb-DF-v2 video subsets AND face-crop boxes, making the
    two models' numbers directly comparable (and skips re-detection).

=====================================================================
DEVIATIONS from the GD-FAS repo defaults (read before trusting results)
=====================================================================
1. Datasets: the repo's O/C/I/M protocol datasets aren't available here;
   we feed LCC-FASD / Celeb-DF-v2 / Asian-Fakes through adapter Dataset
   classes that read straight from /kaggle/input (no folder restructuring,
   nothing copied). Same output dict shape as the repo's own datasets.
2. Single training domain: the repo trains on 3 source domains; our Exp 1
   trains on one dataset (LCC-FASD). With num_domain=1 the repo's
   group-scaling sigmoid slope alpha = log(num_domain)/2 = 0 -> division
   by zero, so we clamp num_domain to >=2 when computing alpha (group-wise
   scaling still operates over the 2 groups live/attack). We also add a
   1e-8 epsilon to the group-loss std (2 groups can have ~equal losses).
3. Eval frame choice: the repo scores ONE RANDOM frame per test video; we
   use the MIDDLE frame for determinism/reproducibility.
4. Eval cadence: the repo evaluates every 10 iterations; with 5 eval
   loaders that's too slow, we evaluate every EVAL_EVERY (40) iterations.
5. Celeb-DF-v2 face crops: reused from the cached index built by the
   G2V2former runs (face-alignment landmark boxes) when available; if no
   cache is found and face_alignment isn't importable, falls back to a
   generous center crop (Celeb-DF is talking-head footage, faces are
   centered) -- flagged in the printout so you know which path ran.
Everything else (CLIP ViT-B/16 backbone, text prompt templates, losses,
group-wise scaling + Gram-Schmidt decomposition, Adam lr 3e-6, batch 16,
StepLR, HTER-at-EER / AUC via the repo's own utils.eval) is the repo's.
"""

# %% CELL 0 -- mount the offline bundle: wheels, CLIP checkpoint, GD-FAS repo
# (bundle built by prepare_gdfas_offline_bundle.py in an internet notebook)
import os
import shutil
import subprocess
import sys
import zipfile

# EDIT to match wherever Kaggle mounted your uploaded bundle dataset. Point
# at EITHER the extracted folder OR the .zip -- both work.
GDFAS_BUNDLE_SRC = "/kaggle/input/datasets/ruwadnaswan/gd-fas-offline-ds/gdfas_offline_bundle"


def _materialize(source, dest_dir):
    """Makes `source` (a directory OR a .zip) available read-ready at a
    stable path. Directories are used in place; zips extract once."""
    if os.path.isdir(source):
        return source
    if os.path.isfile(source) and source.lower().endswith(".zip"):
        if not (os.path.isdir(dest_dir) and os.listdir(dest_dir)):
            os.makedirs(dest_dir, exist_ok=True)
            with zipfile.ZipFile(source, "r") as z:
                z.extractall(dest_dir)
        inner = os.path.join(dest_dir, "gdfas_offline_bundle")
        return inner if os.path.isdir(inner) else dest_dir
    raise FileNotFoundError(f"{source} is neither an existing directory nor a .zip file")


BUNDLE_DIR = _materialize(GDFAS_BUNDLE_SRC, "/kaggle/working/gdfas_bundle")
WHEELS_DIR = os.path.join(BUNDLE_DIR, "wheels")

# Kaggle auto-decompresses .gz files on dataset upload, which strips the
# CLIP tokenizer vocab (bpe_simple_vocab_16e6.txt.gz) down to a plain .txt
# the tokenizer can't open. /kaggle/input is read-only, so: copy the (small)
# repo into /kaggle/working and re-gzip the vocab there.
import gzip

GDFAS_REPO_DIR = "/kaggle/working/GD-FAS"
if not os.path.isdir(GDFAS_REPO_DIR):
    shutil.copytree(os.path.join(BUNDLE_DIR, "GD-FAS"), GDFAS_REPO_DIR)
_vocab_gz = os.path.join(GDFAS_REPO_DIR, "models", "clip", "bpe_simple_vocab_16e6.txt.gz")
_vocab_txt = _vocab_gz[:-3]
if not os.path.isfile(_vocab_gz):
    if not os.path.isfile(_vocab_txt):
        raise FileNotFoundError(f"neither {_vocab_gz} nor {_vocab_txt} exists in the bundle")
    with open(_vocab_txt, "rb") as f_in, gzip.open(_vocab_gz, "wb") as f_out:
        shutil.copyfileobj(f_in, f_out)
    print("re-gzipped tokenizer vocab ->", _vocab_gz)

# 1) install the only non-preinstalled deps, strictly from local wheels
subprocess.check_call([sys.executable, "-m", "pip", "install", "--no-index",
                       f"--find-links={WHEELS_DIR}", "ftfy", "regex"])

# 2) seed ~/.cache/clip so clip.load() finds ViT-B-16.pt (it sha256-checks
# the file and skips the download entirely when it matches)
CLIP_CACHE = os.path.expanduser("~/.cache/clip")
os.makedirs(CLIP_CACHE, exist_ok=True)
_src_ckpt = os.path.join(BUNDLE_DIR, "clip", "ViT-B-16.pt")
_dst_ckpt = os.path.join(CLIP_CACHE, "ViT-B-16.pt")
if not os.path.isfile(_dst_ckpt):
    shutil.copy(_src_ckpt, _dst_ckpt)
print("clip checkpoint ready:", _dst_ckpt)

# 3) the GD-FAS repo on sys.path -- FIRST, so its generic top-level package
# names (models/, data/, utils/) win over anything else lying around
sys.path.insert(0, GDFAS_REPO_DIR)

import torch  # noqa: E402
from models import build_model  # noqa: E402  (GD-FAS repo)
from utils import eval as gdfas_eval  # noqa: E402  (repo's HTER/AUC evaluator)
from data import get_transform  # noqa: E402

print("GD-FAS repo OK:", GDFAS_REPO_DIR)
print("torch:", torch.__version__, "cuda:", torch.cuda.is_available())
if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))


# %% CELL 1 -- config + index LCC-FASD directly from /kaggle/input (no copying)
import random
import re
from collections import defaultdict

import cv2
import numpy as np
from PIL import Image
from torch.utils.data import DataLoader, Dataset
from tqdm import tqdm

SEED = 2025  # repo default
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

CKPT_DIR = "/kaggle/working/ckpts"
CACHE_DIR = "/kaggle/working/cache"
os.makedirs(CKPT_DIR, exist_ok=True)
os.makedirs(CACHE_DIR, exist_ok=True)

DEVICE = "cuda:0"

# ---- training hyperparameters (repo defaults unless noted) ----
BATCH_SIZE = 48          # groups-per-item; images/step = BATCH_SIZE * n_groups * 2 views.
                         # The repo's 16 with 3 domains = 6 groups -> 192 images/step;
                         # we have 2 groups, so 48 restores the same 192 images/step
                         # (which is what lr 3e-6 was tuned against). Trivial for 96GB.
MAX_ITER = 400           # repo default
LR = 3e-6                # repo default (Adam, CLIP backbone)
WD = 1e-6                # repo default
STEP_SIZE = 10           # scheduler.step() every 10 iters, gamma 0.1 (repo)
EVAL_EVERY = 40          # deviation 4: repo evaluates every 10 iters
STAGE2_MAX_ITER = 200    # Exp 2 is adaptation, not from-scratch training
STAGE2_LR = 1e-6         # gentler, to limit forgetting of LCC-FASD
NUM_WORKERS = 4          # data loading (video decode / JPEG reads) is the real
                         # bottleneck on a 96GB GPU, not VRAM; match Kaggle's CPUs

# ---- LCC-FASD (identical indexing to the G2V2former notebook) ----
LCC_INPUT_ROOT = "/kaggle/input/datasets/faber24/lcc-fasd/LCC_FASD"  # EDIT to your mount

LCC_SPLIT_MAP = {
    "LCC_FASD_training": "train",
    "LCC_FASD_development": "test",   # dev merged into "test", same as before
}
LCC_LABEL_MAP = {"real": "live", "spoof": "attack"}


def _lcc_video_id(filename):
    """AA5742_id154_s0_112.png -> AA5742_id154_s0 (drops the frame index)."""
    name = os.path.splitext(filename)[0]
    parts = name.split("_")
    return "_".join(parts[:-1]) if len(parts) > 1 else name


def _frame_sort_key(fname):
    m = re.search(r"_(\d+)\.(png|jpg|jpeg)$", fname, re.IGNORECASE)
    return int(m.group(1)) if m else fname


def index_lcc_fasd(input_root=LCC_INPUT_ROOT, split_map=LCC_SPLIT_MAP, label_map=LCC_LABEL_MAP):
    """Returns {split: [(video_id, [abs frame paths], 'live'|'attack'), ...]}."""
    index = defaultdict(list)
    for split_dir, split_name in split_map.items():
        split_path = os.path.join(input_root, split_dir)
        if not os.path.isdir(split_path):
            print(f"[lcc-fasd] WARNING: {split_path} not found, skipping.")
            continue
        for src_label, dst_label in label_map.items():
            src_path = os.path.join(split_path, src_label)
            if not os.path.isdir(src_path):
                continue
            video_groups = defaultdict(list)
            for img in os.listdir(src_path):
                if img.lower().endswith(".png"):
                    video_groups[_lcc_video_id(img)].append(img)
            for video_id, frames in video_groups.items():
                frames = sorted(frames, key=_frame_sort_key)
                frame_paths = [os.path.join(src_path, f) for f in frames]
                index[split_name].append((f"lcc_{dst_label}_{video_id}", frame_paths, dst_label))
            print(f"[lcc-fasd] {split_name}/{dst_label}: {len(video_groups)} videos indexed (0 bytes copied)")
    return dict(index)


LCC_INDEX = index_lcc_fasd()
print({split: len(samples) for split, samples in LCC_INDEX.items()})


# %% CELL 2 -- Celeb-DF-v2: eval index (official test protocol) + disjoint
# train split. Reuses the G2V2former runs' cached indices when available so
# both models see the exact same videos and face-crop boxes.
CELEBDF_INPUT_ROOT = "/kaggle/input/datasets/reubensuju/celeb-df-v2"  # EDIT to your mount

CELEBDF_EVAL_INDEX_PATH = os.path.join(CACHE_DIR, "celebdf_index.pt")          # same filenames the
CELEBDF_TRAIN_INDEX_PATH = os.path.join(CACHE_DIR, "celebdf_train_index.pt")   # G2V2former runs used
# OPTIONAL: published-cache dataset mount holding the two files above from an
# earlier session ("" to disable). /kaggle/input is read-only, so they seed.
CELEBDF_CACHE_SEED_DIR = ""  # EDIT e.g. "/kaggle/input/datasets/ruwadnaswan/g2v2-caches/cache"

CELEBDF_EVAL_VIDEO_LIMIT = 100    # same caps as the G2V2former runs
CELEBDF_TRAIN_VIDEO_LIMIT = 300
CELEBDF_FRAMES_PER_VIDEO = 16
CELEBDF_CROP_MARGIN = 0.4


def _find_file(root, filename):
    for dirpath, _, filenames in os.walk(root):
        if filename in filenames:
            return os.path.join(dirpath, filename)
    return None


def load_celebdf_protocol(celebdf_root):
    """Official test list -> [(relative_video_path, 'live'|'attack')]."""
    list_path = _find_file(celebdf_root, "List_of_testing_videos.txt")
    items = []
    if list_path is not None:
        with open(list_path) as f:
            for line in f:
                parts = line.strip().split()
                if len(parts) != 2:
                    continue
                _, rel_path = parts
                rel_path = rel_path.replace("\\", "/")
                # Folder name is unambiguous; mirrors of the txt disagree on 0/1 polarity.
                label = "attack" if "synthesis" in rel_path.lower() else "live"
                items.append((rel_path, label))
        print(f"[celebdf] official protocol list: {list_path} ({len(items)} videos)")
    else:
        print("[celebdf] List_of_testing_videos.txt not found -- using ALL videos as eval.")
        for folder, label in [("Celeb-real", "live"), ("YouTube-real", "live"),
                              ("Celeb-synthesis", "attack")]:
            folder_path = os.path.join(celebdf_root, folder)
            if not os.path.isdir(folder_path):
                continue
            for fname in sorted(os.listdir(folder_path)):
                if fname.lower().endswith(".mp4"):
                    items.append((f"{folder}/{fname}", label))
    return items


def load_celebdf_all_items(celebdf_root):
    items = []
    for folder, label in [("Celeb-real", "live"), ("YouTube-real", "live"), ("Celeb-synthesis", "attack")]:
        folder_path = os.path.join(celebdf_root, folder)
        if not os.path.isdir(folder_path):
            continue
        for fname in sorted(os.listdir(folder_path)):
            if fname.lower().endswith(".mp4"):
                items.append((f"{folder}/{fname}", label))
    return items


def load_celebdf_train_items(celebdf_root):
    """All videos EXCLUDING the official test protocol (kept untouched so the
    cross-dataset eval numbers stay comparable across experiments/models)."""
    test_paths = {rel for rel, _ in load_celebdf_protocol(celebdf_root)}
    train_items = [(rel, label) for rel, label in load_celebdf_all_items(celebdf_root)
                   if rel not in test_paths]
    print(f"[celebdf] train split (all minus official test protocol): {len(train_items)} videos")
    return train_items


def _stratified_subset(items, limit, seed=42):
    """Proportional per-label subsample with a fixed seed (plain [:limit]
    risks a class-skewed subset and a misleading HTER/AUC)."""
    by_label = defaultdict(list)
    for item in items:
        by_label[item[1]].append(item)
    rng = random.Random(seed)
    for label in by_label:
        rng.shuffle(by_label[label])
    total = len(items)
    subset = []
    for label, group in by_label.items():
        n = max(1, round(limit * len(group) / total)) if total else 0
        subset.extend(group[:n])
    rng.shuffle(subset)
    return subset[:limit]


def _maybe_face_model():
    """face_alignment is NOT in the GD-FAS bundle; it's only importable if the
    G2V2former offline bundle happens to be mounted too. Optional either way."""
    try:
        import face_alignment
        return face_alignment.FaceAlignment(face_alignment.LandmarksType.THREE_D,
                                            device=DEVICE, flip_input=False)
    except Exception:
        return None


def _detect_box(fa_model, frame_rgb, h, w, margin=CELEBDF_CROP_MARGIN):
    """Square landmark-bbox crop with margin; None -> center crop at decode."""
    if fa_model is None:
        return None
    try:
        preds = fa_model.get_landmarks(frame_rgb)
    except Exception:
        preds = None
    if not preds:
        return None
    xy = preds[0][:, :2].astype(np.float32)
    x_min, y_min = xy.min(axis=0)
    x_max, y_max = xy.max(axis=0)
    bw, bh = max(x_max - x_min, 1.0), max(y_max - y_min, 1.0)
    cx, cy = (x_min + x_max) / 2, (y_min + y_max) / 2
    side = max(bw, bh) * (1.0 + margin)
    x0, x1 = int(max(0, cx - side / 2)), int(min(w, cx + side / 2))
    y0, y1 = int(max(0, cy - side / 2)), int(min(h, cy + side / 2))
    if x1 <= x0 or y1 <= y0:
        return None
    return (x0, y0, x1, y1)


def build_celebdf_index(celebdf_root, items, cache_path, seed_path=None,
                        frames_per_video=CELEBDF_FRAMES_PER_VIDEO):
    """[{'video_path', 'label' (1=live/0=attack), 'frames': [{'frame_idx','box'},...]}]
    -- metadata only, no pixels written. Same shape as the G2V2former caches
    (whose entries carry extra keys like 'lmk_crop'; we just ignore those)."""
    if os.path.exists(cache_path):
        index = torch.load(cache_path, weights_only=False)
        print(f"[celebdf] loaded cached index: {len(index)} videos from {cache_path}")
        return index
    if seed_path and os.path.exists(seed_path):
        index = torch.load(seed_path, weights_only=False)
        torch.save(index, cache_path)
        print(f"[celebdf] seeded cached index: {len(index)} videos from {seed_path}")
        return index

    fa_model = _maybe_face_model()
    print(f"[celebdf] building index for {len(items)} videos "
          f"({'landmark boxes' if fa_model else 'CENTER-CROP FALLBACK (deviation 5)'})")
    index = []
    for rel_path, label in tqdm(items, desc="celebdf: indexing (metadata only)"):
        video_path = os.path.join(celebdf_root, rel_path)
        if not os.path.isfile(video_path):
            continue
        cap = cv2.VideoCapture(video_path)
        total = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
        if total <= 0:
            cap.release()
            continue
        target_set = set(np.linspace(0, total - 1, min(frames_per_video, total))
                         .round().astype(int).tolist())
        frame_meta, i = [], 0
        while len(frame_meta) < len(target_set):
            ok = cap.grab()
            if not ok:
                break
            if i in target_set:
                ok, frame = cap.retrieve()
                if ok:
                    h, w = frame.shape[:2]
                    box = _detect_box(fa_model, cv2.cvtColor(frame, cv2.COLOR_BGR2RGB), h, w)
                    frame_meta.append({"frame_idx": i, "box": box})
            i += 1
        cap.release()
        if frame_meta:
            index.append({"video_path": video_path, "label": int(label == "live"),
                          "frames": frame_meta})
    torch.save(index, cache_path)
    print(f"[celebdf] built + cached {len(index)} videos -> {cache_path}")
    return index


_seed = lambda name: os.path.join(CELEBDF_CACHE_SEED_DIR, name) if CELEBDF_CACHE_SEED_DIR else None

celebdf_eval_index = build_celebdf_index(
    CELEBDF_INPUT_ROOT,
    _stratified_subset(load_celebdf_protocol(CELEBDF_INPUT_ROOT), CELEBDF_EVAL_VIDEO_LIMIT),
    CELEBDF_EVAL_INDEX_PATH, _seed("celebdf_index.pt"))

celebdf_train_index = build_celebdf_index(
    CELEBDF_INPUT_ROOT,
    _stratified_subset(load_celebdf_train_items(CELEBDF_INPUT_ROOT), CELEBDF_TRAIN_VIDEO_LIMIT),
    CELEBDF_TRAIN_INDEX_PATH, _seed("celebdf_train_index.pt"))

print(f"celebdf eval videos: {len(celebdf_eval_index)}  train videos: {len(celebdf_train_index)}")


# %% CELL 3 -- index the Asian-Fakes stills (single frames, not clips).
# NOTE: on-disk folder names contain "msib_*" (that's just the data layout);
# every DISPLAY name below is msib-free per the reporting convention.
ASIAN_FAKES_ROOT = "/kaggle/input/notebooks/raihanzahin/prep-sbi-extra/sbi_faces_extra/images"  # EDIT

ASIAN_SUBGROUPS = {"msib_asian": "asian", "msib_non_asian": "non-asian"}  # folder -> display
ASIAN_LABEL_MAP = {"real": "live", "sbi": "attack"}
ASIAN_IMG_EXTS = (".png", ".jpg", ".jpeg", ".bmp", ".webp")


def index_asian_fakes(root=ASIAN_FAKES_ROOT):
    """{'all'|'asian'|'non-asian': [(sample_id, [abs path], 'live'|'attack'), ...]}"""
    out = {"all": []}
    for folder, display in ASIAN_SUBGROUPS.items():
        sub_path = os.path.join(root, folder)
        out[display] = []
        if not os.path.isdir(sub_path):
            print(f"[asian-fakes] WARNING: {sub_path} not found, skipping.")
            continue
        for src_label, dst_label in ASIAN_LABEL_MAP.items():
            src_path = os.path.join(sub_path, src_label)
            if not os.path.isdir(src_path):
                print(f"[asian-fakes] WARNING: {src_path} not found, skipping.")
                continue
            n = 0
            for fname in sorted(os.listdir(src_path)):
                if fname.lower().endswith(ASIAN_IMG_EXTS):
                    entry = (f"asian_{display}_{dst_label}_{os.path.splitext(fname)[0]}",
                             [os.path.join(src_path, fname)], dst_label)
                    out[display].append(entry)
                    out["all"].append(entry)
                    n += 1
            print(f"[asian-fakes] {display}/{dst_label}: {n} images indexed (0 bytes copied)")
    print(f"[asian-fakes] total: {len(out['all'])} images "
          f"({len(out['asian'])} asian, {len(out['non-asian'])} non-asian)")
    return out


ASIAN_INDEX = index_asian_fakes()


# %% CELL 4 -- adapter datasets feeding the GD-FAS training/eval interface.
# Same output dicts as the repo's BalanceFaceDataset / FaceDataset, but
# reading LCC frame folders, Celeb-DF mp4s, and Asian-Fakes stills straight
# from /kaggle/input.
train_transform, test_transform = get_transform("clip")  # repo's own 224px CLIP transforms


def _frames_sample(sample_id, frame_paths, label_str, domain):
    return {"kind": "frames", "id": sample_id, "frame_paths": frame_paths,
            "label": int(label_str == "live"), "domain": domain}


def _video_sample(entry, domain):
    return {"kind": "video", "id": os.path.basename(entry["video_path"]),
            "video_path": entry["video_path"], "frames": entry["frames"],
            "label": int(entry["label"]), "domain": domain}


def _center_box(h, w):
    side = int(min(h, w) * 0.8)
    y0, x0 = (h - side) // 2, (w - side) // 2
    return (x0, y0, x0 + side, y0 + side)


def _decode_video_frame(video_path, frame_idx, box):
    cap = cv2.VideoCapture(video_path)
    cap.set(cv2.CAP_PROP_POS_FRAMES, frame_idx)
    ok, frame = cap.read()
    cap.release()
    if not ok or frame is None:
        return Image.new("RGB", (224, 224))
    h, w = frame.shape[:2]
    x0, y0, x1, y1 = box if box is not None else _center_box(h, w)
    crop = frame[max(0, y0):min(h, y1), max(0, x0):min(w, x1)]
    if crop.size == 0:
        crop = frame
    return Image.fromarray(cv2.cvtColor(crop, cv2.COLOR_BGR2RGB))


def load_sample_pil(sample, which="random"):
    """which: 'random' (training, like the repo) or 'middle' (deterministic eval)."""
    if sample["kind"] == "frames":
        paths = sample["frame_paths"]
        idx = np.random.randint(len(paths)) if which == "random" else len(paths) // 2
        return Image.open(paths[idx]).convert("RGB")
    metas = sample["frames"]
    meta = metas[np.random.randint(len(metas))] if which == "random" else metas[len(metas) // 2]
    return _decode_video_frame(sample["video_path"], meta["frame_idx"], meta.get("box"))


class GDFASBalancedDataset(Dataset):
    """Mirrors the repo's BalanceFaceDataset: each item carries one sample per
    (dataset x live/attack) group, each as two augmented views, so every
    training step is class- and domain-balanced. len = max_iter * batch_size,
    exactly like the repo, so the DataLoader yields max_iter batches."""

    def __init__(self, groups, transform, max_iter, batch_size):
        self.groups = groups          # {group_key: [sample, ...]}
        self.transform = transform
        self.length = max_iter * batch_size
        for key, samples in groups.items():
            assert samples, f"empty training group {key}"

    def __len__(self):
        return self.length

    def __getitem__(self, idx):
        out = {}
        for key, samples in self.groups.items():
            s = samples[np.random.randint(len(samples))]
            img = load_sample_pil(s, "random")
            out[key] = {
                "image_x_v1": self.transform(img),
                "image_x_v2": self.transform(img),
                "label": s["label"],
                "domain": s["domain"],
            }
        return out


class GDFASEvalDataset(Dataset):
    """Mirrors the repo's FaceDataset: one frame per video/still. Deviation 3:
    middle frame instead of a random one, for run-to-run reproducibility."""

    def __init__(self, samples, transform):
        self.samples = samples
        self.transform = transform

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        s = self.samples[idx]
        return {"image_x": self.transform(load_sample_pil(s, "middle")),
                "label": s["label"]}


# ---- concrete sample lists ----
lcc_train_samples = [_frames_sample(vid, paths, lab, domain=0) for vid, paths, lab in LCC_INDEX["train"]]
lcc_test_samples = [_frames_sample(vid, paths, lab, domain=0) for vid, paths, lab in LCC_INDEX["test"]]
celebdf_eval_samples = [_video_sample(e, domain=0) for e in celebdf_eval_index]
celebdf_train_samples = [_video_sample(e, domain=0) for e in celebdf_train_index]
asian_samples = {k: [_frames_sample(sid, paths, lab, domain=0) for sid, paths, lab in v]
                 for k, v in ASIAN_INDEX.items()}


def make_eval_loader(samples):
    return DataLoader(GDFASEvalDataset(samples, test_transform), batch_size=256,
                      shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)


EVAL_LOADERS = {
    "LCC-FASD (test)": make_eval_loader(lcc_test_samples),
    "CelebDF-v2 (test)": make_eval_loader(celebdf_eval_samples),
    "Asian-Fakes (all)": make_eval_loader(asian_samples["all"]),
    "Asian-Fakes (asian)": make_eval_loader(asian_samples["asian"]),
    "Asian-Fakes (non-asian)": make_eval_loader(asian_samples["non-asian"]),
}


def make_train_loader(groups, max_iter):
    ds = GDFASBalancedDataset(groups, train_transform, max_iter, BATCH_SIZE)
    return DataLoader(ds, batch_size=BATCH_SIZE, shuffle=False,
                      num_workers=NUM_WORKERS, pin_memory=True)


def split_by_class(samples, prefix):
    return {f"{prefix}-attack": [s for s in samples if s["label"] == 0],
            f"{prefix}-live": [s for s in samples if s["label"] == 1]}


# %% CELL 5 -- model builder + training loop (adapted from GD-FAS.py main();
# same losses/optimizer/scheduler, generalized to a dict of eval loaders and
# to state_dict checkpointing).
import types
from types import SimpleNamespace


def _safe_group_wise_scaling(self, logits, labels):
    """Repo's group_wise_scaling with a std epsilon (deviation 2: with only
    2 groups the per-group losses can be near-identical -> std ~ 0 -> nan)."""
    sizes = [l.size(0) for l in labels]
    losses = [torch.nn.functional.cross_entropy(logit, label) * size / sum(sizes)
              for logit, label, size in zip(logits, labels, sizes)]
    n_losses = np.array([loss.item() for loss in losses])
    norm_losses = (n_losses - n_losses.mean()) / (n_losses.std() + 1e-8)
    return [l * self.scaling_estimator(n) for l, n in zip(losses, norm_losses)], losses


def build_gdfas_model(num_train_domains=1):
    args = SimpleNamespace(
        backbone="clip", silence=True, gs=True,
        beta=1.5, temperature=0.1, params=[1.0, 0.8, 0.1, 1.0],
        # deviation 2: alpha = log(num_domain)/2 divides by zero at 1 domain
        num_domain=max(2, num_train_domains),
    )
    model = build_model(args).cuda()
    model.group_wise_scaling = types.MethodType(_safe_group_wise_scaling, model)
    return model


@torch.no_grad()
def evaluate_loader(model, loader):
    """Repo's scoring: softmax over image-text similarity logits, prob of the
    'real' template = liveness score, then utils.eval -> HTER at EER / AUC."""
    model.eval()
    lines = []
    for batch in loader:
        images = batch["image_x"].cuda(non_blocking=True)
        logits_per_image, _ = model(images)
        probs = torch.nn.functional.softmax(logits_per_image, dim=1)
        for prob, label in zip(probs, batch["label"]):
            lines.append(f"{prob[1].item()} {label.item()}\n")
    _, _, HTER, auc_score, _, _, _, _, _ = gdfas_eval(lines)
    return HTER[0] * 100.0, auc_score * 100.0   # % convention, matching lets-see-2


def run_full_eval(model, loaders, tag):
    print(f"\n===== Evaluating [{tag}] on all datasets =====")
    results = {}
    for name, loader in loaders.items():
        hter, auc_val = evaluate_loader(model, loader)
        results[name] = {"HTER": hter, "AUC": auc_val}
        print(f"[{tag}] {name:24s} HTER={hter:6.2f}%  AUC={auc_val:6.2f}%")
    return results


def train_gdfas(model, train_loader, eval_loaders, primary_key, max_iter, lr,
                tag, eval_every=EVAL_EVERY):
    """GD-FAS.py's loop: cat the two views of every balanced group, one
    optimizer step per batch, StepLR stepping every 10 iters (=> lr x0.1
    every 100 iters), best checkpoint tracked by HTER on `primary_key`."""
    import torch.optim as optim
    optimizer = optim.Adam(model.parameters(), lr=lr, weight_decay=WD)
    scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=STEP_SIZE, gamma=0.1)

    best_hter, best_path = float("inf"), os.path.join(CKPT_DIR, f"gdfas_{tag}_best.pt")
    history = []
    pbar = tqdm(train_loader, total=max_iter, desc=f"[{tag}] train")
    for it, batch_samples in enumerate(pbar):
        model.train()
        optimizer.zero_grad()

        image_x_v1 = torch.cat([batch_samples[key]["image_x_v1"] for key in batch_samples])
        image_x_v2 = torch.cat([batch_samples[key]["image_x_v2"] for key in batch_samples])
        images = torch.cat([image_x_v1, image_x_v2]).cuda(non_blocking=True)
        labels = torch.cat([batch_samples[key]["label"] for key in batch_samples]).repeat(2).cuda()
        domains = torch.cat([batch_samples[key]["domain"] for key in batch_samples]).repeat(2).cuda()

        loss = model.compute_loss(images, labels, domains)
        loss.backward()
        optimizer.step()
        pbar.set_postfix(loss=f"{loss.item():.4f}")

        if (it % STEP_SIZE == 0) and (it != 0):
            scheduler.step()

        if ((it + 1) % eval_every == 0) or (it + 1 == max_iter):
            print(f"\n[{tag}] iter {it + 1}: {model.loss_reset()['loss']}")
            results = run_full_eval(model, eval_loaders, f"{tag} iter {it + 1}")
            history.append({"iter": it + 1, "results": results})
            if results[primary_key]["HTER"] <= best_hter:
                best_hter = results[primary_key]["HTER"]
                torch.save(model.state_dict(), best_path)
                print(f"  -> new best [{tag}] checkpoint saved "
                      f"(HTER on '{primary_key}' = {best_hter:.2f}%)")
    print(f"[{tag}] best HTER on '{primary_key}': {best_hter:.2f}%  ({best_path})")
    return best_path, history


ALL_RESULTS = {}


# %% CELL 6 -- EXPERIMENT 1: train GD-FAS on LCC-FASD only, test on all 3
# datasets. Groups = LCC live / LCC attack (single domain -- deviation 2).
exp1_groups = split_by_class(lcc_train_samples, "LCC")
exp1_model = build_gdfas_model(num_train_domains=1)
exp1_best_ckpt, exp1_history = train_gdfas(
    exp1_model, make_train_loader(exp1_groups, MAX_ITER), EVAL_LOADERS,
    primary_key="LCC-FASD (test)", max_iter=MAX_ITER, lr=LR, tag="lcc")

# Final scoring off the saved BEST checkpoint, same convention as before.
exp1_best_model = build_gdfas_model(num_train_domains=1)
exp1_best_model.load_state_dict(torch.load(exp1_best_ckpt, map_location=DEVICE))
ALL_RESULTS["Exp1: train=LCC-FASD"] = run_full_eval(
    exp1_best_model, EVAL_LOADERS, tag="Exp1: train=LCC-FASD")


# %% CELL 7 -- EXPERIMENT 2: continue from the Exp-1 best checkpoint,
# finetune on the Celeb-DF-v2 TRAIN split (disjoint from the official test
# protocol), test on the identical EVAL_LOADERS.
assert os.path.exists(exp1_best_ckpt), "run CELL 6 first"

exp2_groups = split_by_class(celebdf_train_samples, "CelebDF")
exp2_model = build_gdfas_model(num_train_domains=1)
exp2_model.load_state_dict(torch.load(exp1_best_ckpt, map_location=DEVICE))
exp2_best_ckpt, exp2_history = train_gdfas(
    exp2_model, make_train_loader(exp2_groups, STAGE2_MAX_ITER), EVAL_LOADERS,
    primary_key="CelebDF-v2 (test)",   # track best by the dataset being finetuned on
    max_iter=STAGE2_MAX_ITER, lr=STAGE2_LR, tag="lcc_then_celebdf", eval_every=EVAL_EVERY // 2)

exp2_best_model = build_gdfas_model(num_train_domains=1)
exp2_best_model.load_state_dict(torch.load(exp2_best_ckpt, map_location=DEVICE))
ALL_RESULTS["Exp2: train=LCC-FASD -> finetune=Celeb-DF-v2"] = run_full_eval(
    exp2_best_model, EVAL_LOADERS, tag="Exp2: train=LCC-FASD -> finetune=Celeb-DF-v2")


# %% CELL 8 -- results table + grouped bar charts (HTER, AUC) + CSV, same
# format as the G2V2former comparison so the two models line up side by side.
import matplotlib.pyplot as plt
import pandas as pd


def results_to_dataframe(all_results):
    rows = []
    for exp_name, ds_results in all_results.items():
        for ds_name, metrics in ds_results.items():
            rows.append({"experiment": exp_name, "dataset": ds_name,
                         "HTER (%)": metrics["HTER"], "AUC (%)": metrics["AUC"]})
    return pd.DataFrame(rows)


results_df = results_to_dataframe(ALL_RESULTS)
pivot = results_df.pivot(index="dataset", columns="experiment",
                         values=["HTER (%)", "AUC (%)"]).round(2)
print(pivot)

csv_path = os.path.join(CKPT_DIR, "gdfas_experiment_comparison.csv")
results_df.to_csv(csv_path, index=False)
print(f"\nSaved results table -> {csv_path}")


def plot_metric_comparison(df, metric, save_path=None):
    datasets = df["dataset"].unique().tolist()
    experiments = df["experiment"].unique().tolist()
    x = np.arange(len(datasets))
    width = 0.8 / max(1, len(experiments))

    fig, ax = plt.subplots(figsize=(1.8 * len(datasets) + 3, 5))
    for i, exp in enumerate(experiments):
        vals = []
        for d in datasets:
            row = df[(df.dataset == d) & (df.experiment == exp)]
            vals.append(row[metric].values[0] if len(row) else np.nan)
        offset = i * width - width * (len(experiments) - 1) / 2
        bars = ax.bar(x + offset, vals, width, label=exp)
        ax.bar_label(bars, fmt="%.1f", fontsize=8, padding=2)

    ax.set_xticks(x)
    ax.set_xticklabels(datasets, rotation=20, ha="right")
    ax.set_ylabel(metric)
    ax.set_title(f"GD-FAS: {metric} by dataset and experiment")
    ax.legend(fontsize=8)
    ax.grid(axis="y", alpha=0.3)
    fig.tight_layout()
    if save_path:
        fig.savefig(save_path, dpi=150)
        print(f"Saved -> {save_path}")
    plt.show()


plot_metric_comparison(results_df, "HTER (%)", os.path.join(CKPT_DIR, "gdfas_hter_comparison.png"))
plot_metric_comparison(results_df, "AUC (%)", os.path.join(CKPT_DIR, "gdfas_auc_comparison.png"))
```
