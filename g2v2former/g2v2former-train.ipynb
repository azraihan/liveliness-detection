```python
"""
train_g2v2former.py
====================
G2V2former ("Graph Guided Video Vision Transformer for Face Anti-Spoofing")
-- unofficial, from-scratch implementation (arXiv:2408.07675), since the
authors have not released code.

Run this in the OFFLINE, no-internet, 96GB-VRAM Kaggle notebook. Every cell
below is self-contained (marked `# %% CELL n`) so you can paste each one
into its own notebook cell and run top to bottom -- or run the file as a
plain script.

Inputs this notebook expects to be attached (as Kaggle "Add Data" inputs,
no internet needed once attached):
  - The offline bundle produced by prepare_offline_bundle.py
    (offline_packages.zip + face_alignment_hub.zip), run separately in an
    internet-enabled notebook. See that file's docstring.
  - LCC-FASD (faber24/lcc-fasd)
  - Celeb-DF-v2 (reubensuju/celeb-df-v2) -- used for cross-dataset,
    deepfake-domain evaluation after every fine-tuning epoch.

Nothing from those datasets is copied into /kaggle/working. Frames and
video files are always read directly from /kaggle/input at __getitem__
time; the only things written to /kaggle/working are small metadata caches
(landmark coordinates, a Celeb-DF-v2 frame/crop index) and model
checkpoints -- see fix J below.

=====================================================================
DEVIATIONS FROM THE PAPER (unchanged from the original draft -- read before
trusting results)
=====================================================================
The paper does not release hyperparameters, exact layer widths, the
landmark graph topology, or several implementation details. Where the
paper is silent, this script makes an explicit, documented choice. With
one dataset instead of their nine, the paper's cross-dataset/LOO protocol
is impossible to reproduce here -- this script does a plain train/test
split on LCC-FASD (presentation-attack detection: print/replay), reports
HTER/AUC on that held-out split, AND (new, see below) also reports HTER/AUC
on Celeb-DF-v2, which is a DIFFERENT attack family (deepfakes, not
print/replay). That cross-dataset number should be read as "how well does
a PAD-trained model generalize to deepfakes", not as a same-distribution
result, and it is normal/expected for it to look much worse than the
in-domain LCC-FASD number, especially early in training.

1. Landmark extractor: paper uses 3DDFA-V2. We use `face-alignment`
   (1adrianb/face-alignment) 3D landmarks -- functionally similar, not
   bit-identical.
2. Landmark graph edges (68-point adjacency) and max-hop=2: not specified
   by the paper; we use standard iBUG-68 face-mesh connectivity.
3. Hyper-node topology: fully connected (hop=1) to every landmark, acting
   like a CLS token for the graph stream (undocumented by the paper;
   without this the hyper-node would be isolated by hop-distance masking).
4. Model widths/depths/heads, LR, batch size, optimizer betas, drop-path,
   PCL tau, CE/PCL loss weight: mostly unspecified by the paper. We keep
   what they *did* specify (Adam, cosine decay + warmup, base LR 5e-4
   pretrain / 2e-4 finetune, 12 encoder / 4 decoder layers, patch size 16,
   8 frames, 90% masking) and pick ViT-S-scale defaults elsewhere.
5. MAE pretraining is the SimMIM-style variant (masked positions replaced
   by a learned mask token, sequence length fixed) rather than token
   removal, so the fine-tuning blocks can be reused unmodified.
6. Graph-guidance bias (Eq. 15-16) is averaged across attention heads
   before being broadcast into the vision stream (paper doesn't specify
   head reconciliation between the two streams).
7. Data augmentation is simplified (flip + small rotation).
8. LCC-FASD frame folders are a pseudo-clip stand-in for genuine video;
   temporal continuity across frames in a folder is not guaranteed.

=====================================================================
BUG FIXES applied in this revision (the original draft had a NaN-loss bug)
=====================================================================
A. ROOT CAUSE OF THE NaN LOSS: `scatter_graph_bias_to_vision` injected the
   graph stream's raw (pre-softmax, unbounded) attention logits into the
   vision stream's attention logits via `scatter_add_` -- i.e. a SUM, not
   an average, over every colliding landmark-pair that maps onto the same
   (patch_i, patch_j) cell. Since 68 landmarks concentrate into a 14x14
   patch grid, many landmark pairs collide into the same cell (e.g. every
   pair among the ~5 landmarks around an eye corner lands in one patch),
   and the summed bias magnitude scales with that (uncontrolled,
   data-dependent) collision count instead of the natural O(1-10)
   attention-logit scale. Under fp16 autocast this blows past the fp16 max
   (~65504) within a few steps -> `inf` -> `inf - inf = nan` in the
   softmax. FIXED by turning the scatter into a scatter-MEAN (divide by
   contribution count) and clamping to +-`CFG.guide_bias_clip`.
B. All masking (`kronecker_temporal_mask`, `TopoSpatialAttention` hop bias)
   now uses a large finite constant (`NEG_INF = -1e4`) instead of literal
   `-inf`, avoiding any possibility of `-inf + -inf` / `exp(-inf)` edge
   cases interacting badly with autocast's dtype promotions.
C. AMP now defaults to **bf16** (`CFG.amp_dtype = "bf16"`), which has the
   same exponent range as fp32 and therefore doesn't overflow the way fp16
   does -- the single most effective guard against this class of NaN,
   and a good fit given the 96GB-class GPU this is meant to run on. Falls
   back to fp16 + GradScaler automatically if bf16 isn't supported.
D. Added a linear LR warmup before the cosine decay (the paper says it
   uses warmup; the original draft silently dropped it and started at
   full LR from step 0, a classic cause of early-training divergence for
   a from-scratch 24-layer transformer).
E. Added an explicit non-finite-loss guard in every training loop: a
   non-finite loss skips that step (zero-grad, no optimizer step, no
   scheduler step) instead of silently corrupting the model; too many
   consecutive skips raises a clear error instead of burning the rest of
   the run on a diverged model.
F. Vectorized `scatter_graph_bias_to_vision` (was a Python for-loop over
   the batch) and `VisionMAE.random_mask` / `GraphMAE.random_mask` (were
   Python double for-loops over batch x time) -- correctness-preserving,
   meaningful speedup on a big GPU with a bigger batch size.
G. Deprecated `torch.cuda.amp.autocast` / `torch.cuda.amp.GradScaler` ->
   `torch.amp.autocast('cuda', ...)` / `torch.amp.GradScaler('cuda', ...)`.
   `torch.load(...)` calls for model checkpoints now pass `weights_only=True`.
H. Cleaned up dead code (`probs.argmax_placeholder = ...`).
I. Celeb-DF-v2 cross-dataset (deepfake) evaluation after every fine-tuning
   epoch, with its own best-checkpoint tracking.
J. STOPPED DUPLICATING INPUT DATA INTO /kaggle/working. The original draft
   (and an earlier revision of this file) physically copied every LCC-FASD
   frame into /kaggle/working/dataset/LCC_FASD, and extracted+saved a
   cropped PNG per sampled Celeb-DF-v2 frame -- multiple GB of copies of
   data that already lives, read-only, under /kaggle/input. Now:
     - LCC-FASD: `index_lcc_fasd()` groups the flat per-frame files into
       per-video sample lists *in memory* (just `os.listdir` + grouping by
       the video id encoded in the filename); `FASVideoDataset` reads each
       frame straight from its original /kaggle/input path.
     - Celeb-DF-v2: `build_celebdf_index()` walks each selected video once,
       running face detection only to record a crop box + landmarks per
       sampled frame (a few hundred bytes), never the pixels themselves.
       `CelebDFVideoDataset` re-opens the source .mp4 directly from
       /kaggle/input at __getitem__ time, decodes just the needed frames,
       and applies the cached crop box.
   The only things this script writes to /kaggle/working are: the landmark
   cache (`CFG.landmark_cache_path`, one consolidated file -- can't write
   a sibling *_lmk.npy next to each frame when the frame lives under
   read-only /kaggle/input), the Celeb-DF-v2 frame/crop index
   (`CFG.celebdf_index_path`, similarly tiny), and model checkpoints under
   `CFG.ckpt_dir`. Both caches are cheap (tens of MB, not GB) and
   resumable -- safe to re-run a cell if the kernel restarts.
=====================================================================
"""

# %% CELL 0 -- mount the offline package bundle + face-alignment hub cache
# (built by prepare_offline_bundle.py in a separate, internet-enabled notebook)
#
# Works whether your uploaded dataset stores these as .zip files or as
# already-extracted folders (Kaggle datasets can be published either way,
# and prepare_offline_bundle.py's own zips extract to this same shape --
# this handles both without you needing to know which).
import os
import shutil
import sys
import zipfile

# EDIT these to match wherever Kaggle mounted your uploaded dataset. Point
# at EITHER the .zip file OR the already-extracted folder -- both work.
OFFLINE_PACKAGES_SRC = "/kaggle/input/datasets/ruwadnaswan/package-df-msib/offline_packages"
FACE_ALIGNMENT_HUB_SRC = "/kaggle/input/datasets/ruwadnaswan/package-df-msib/face_alignment_hub"


def _materialize(source, dest_dir):
    """Makes `source` (a directory OR a .zip file) available, read-ready, at
    `dest_dir`. An input directory is used in place -- no copy needed, read
    access is all sys.path requires. A .zip gets extracted into dest_dir
    once (skipped on re-run if dest_dir is already populated)."""
    if os.path.isdir(source):
        return source
    if os.path.isfile(source) and source.lower().endswith(".zip"):
        if not (os.path.isdir(dest_dir) and os.listdir(dest_dir)):
            os.makedirs(dest_dir, exist_ok=True)
            with zipfile.ZipFile(source, "r") as z:
                z.extractall(dest_dir)
        return dest_dir
    raise FileNotFoundError(f"{source} is neither an existing directory nor a .zip file")


SITE_PACKAGES_DIR = _materialize(OFFLINE_PACKAGES_SRC, "/kaggle/working/offline_site_packages")
sys.path.insert(0, SITE_PACKAGES_DIR)

# torch.hub sometimes touches its cache dir (lock files etc.) even when
# every checkpoint it needs is already present, so give it a writable copy
# rather than pointing straight at read-only /kaggle/input.
HUB_DIR = "/kaggle/working/hub"
if not (os.path.isdir(HUB_DIR) and os.listdir(HUB_DIR)):
    if os.path.isdir(FACE_ALIGNMENT_HUB_SRC):
        shutil.copytree(FACE_ALIGNMENT_HUB_SRC, HUB_DIR, dirs_exist_ok=True)
    else:
        _materialize(FACE_ALIGNMENT_HUB_SRC, HUB_DIR)

import torch  # noqa: E402
torch.hub.set_dir(HUB_DIR)

import face_alignment  # noqa: E402  (sanity check: importable from the offline bundle)
print("face_alignment OK, version:", getattr(face_alignment, "__version__", "unknown"))
print("torch:", torch.__version__, "cuda available:", torch.cuda.is_available())
if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0),
          "bf16 supported:", torch.cuda.is_bf16_supported())


# %% CELL 1 -- index LCC-FASD directly from /kaggle/input (no copying)
# LCC-FASD's frames stay exactly where Kaggle mounted them (/kaggle/input is
# read-only anyway); this just groups the flat per-frame files into
# per-video sample lists in memory. Nothing gets written to /kaggle/working.
import re
from collections import defaultdict
from tqdm import tqdm

LCC_INPUT_ROOT = "/kaggle/input/datasets/faber24/lcc-fasd/LCC_FASD"          # EDIT to match your attached dataset path

LCC_SPLIT_MAP = {
    "LCC_FASD_training": "train",
    "LCC_FASD_development": "test",   # merge dev into "test"; swap in LCC_FASD_evaluation if you'd rather hold that out
}
LCC_LABEL_MAP = {"real": "live", "spoof": "attack"}


def _lcc_video_id(filename):
    """AA5742_id154_s0_112.png -> AA5742_id154_s0 (drops the frame index)."""
    name = os.path.splitext(filename)[0]
    parts = name.split("_")
    return "_".join(parts[:-1]) if len(parts) > 1 else name


def _frame_sort_key(fname: str):
    m = re.search(r"_(\d+)\.(png|jpg|jpeg)$", fname, re.IGNORECASE)
    return int(m.group(1)) if m else fname


def index_lcc_fasd(input_root=LCC_INPUT_ROOT, split_map=LCC_SPLIT_MAP, label_map=LCC_LABEL_MAP):
    """Returns {split_name: [(video_id, [abs frame paths under /kaggle/input], label_str), ...]}."""
    index = defaultdict(list)
    for split_dir, split_name in split_map.items():
        split_path = os.path.join(input_root, split_dir)
        if not os.path.isdir(split_path):
            print(f"[lcc-fasd] WARNING: {split_path} not found, skipping.")
            continue
        for src_label, dst_label in label_map.items():
            src_path = os.path.join(split_path, src_label)
            if not os.path.isdir(src_path):
                continue
            video_groups = defaultdict(list)
            for img in os.listdir(src_path):
                if img.lower().endswith(".png"):
                    video_groups[_lcc_video_id(img)].append(img)
            for video_id, frames in video_groups.items():
                frames = sorted(frames, key=_frame_sort_key)
                frame_paths = [os.path.join(src_path, f) for f in frames]
                index[split_name].append((f"lcc_{dst_label}_{video_id}", frame_paths, dst_label))
            print(f"[lcc-fasd] {split_name}/{dst_label}: {len(video_groups)} videos indexed (0 bytes copied)")
    return dict(index)


LCC_INDEX = index_lcc_fasd()
print({split: len(samples) for split, samples in LCC_INDEX.items()})


# %% CELL 2 -- imports & global config
import functools
import math
import random
import numpy as np
import cv2
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from collections import deque
from typing import List, Tuple

try:
    from sklearn.metrics import roc_curve, auc as sk_auc
except ImportError:
    roc_curve = sk_auc = None

torch.set_float32_matmul_precision("high")

# Large finite value used for attention masking instead of literal -inf
# (see fix B in the module docstring).
NEG_INF = -1e4


class CFG:
    # ---- data ----
    img_size = 224
    patch_size = 16
    num_frames = 8
    num_landmarks = 68
    max_hop = 2
    num_freq_bands = 6

    # ---- caches (small, metadata-only -- see fix J; nothing is copied
    # into /kaggle/working besides these) ----
    landmark_cache_path = "/kaggle/working/cache/lcc_landmarks.pt"
    celebdf_index_path = "/kaggle/working/cache/celebdf_index.pt"

    # Optional: point these at a cache you built on an earlier run and then
    # published as a Kaggle Dataset (upload /kaggle/working/cache/*.pt),
    # mounted here as an input. /kaggle/input has a much bigger quota
    # (~200GB) than /kaggle/working AND persists across sessions, so once
    # you've paid for landmark extraction / Celeb-DF-v2 indexing once,
    # pointing here skips re-running face-alignment over the whole dataset
    # on every restart. Leave as None on your very first run -- there's
    # nothing published yet.
    landmark_cache_input_path = None    # e.g. "/kaggle/input/g2v2former-cache/lcc_landmarks.pt"
    celebdf_index_input_path = None     # e.g. "/kaggle/input/g2v2former-cache/celebdf_index.pt"

    # ---- model dims (paper-unspecified -> ViT-S-ish, single-GPU friendly) ----
    vis_dim = 384
    vis_depth = 12
    vis_heads = 6
    graph_dim = 192
    graph_depth = 12
    graph_heads = 6
    mlp_ratio = 4.0
    num_classes = 2

    # ---- graph -> vision guidance bias (fix A: mean + clamp, was unclamped sum) ----
    guide_bias_clip = 8.0

    # ---- MAE pretraining (Stage 1, run independently per stream) ----
    do_pretrain = True
    pretrain_mask_ratio = 0.90          # paper-specified
    pretrain_epochs_vis = 8
    pretrain_epochs_graph = 8
    pretrain_lr = 5e-4                  # paper-specified
    pretrain_batch_size = 32            # raised from 16 -- you have a 96GB GPU; push higher if headroom allows
    decoder_depth = 4                   # paper-specified ("asymmetric... 4 layers decoder")
    decoder_dim_vis = 192
    decoder_dim_graph = 96

    # ---- fine-tuning (Stage 2) ----
    finetune_epochs = 20
    finetune_lr = 2e-4                  # paper-specified
    finetune_batch_size = 32            # raised from 16; raise further if VRAM allows
    pcl_tau = 0.07
    pcl_weight = 1.0
    grad_clip = 1.0
    warmup_ratio = 0.05                 # fix D: linear warmup before cosine decay

    # ---- AMP / numerical stability (fix C) ----
    amp = True
    amp_dtype = "bf16"                  # "bf16" (recommended on 80GB+ Ampere/Hopper/Blackwell) or "fp16"
    max_nan_skips = 20                  # fix E: abort if this many consecutive steps produce a non-finite loss

    # ---- Celeb-DF-v2 cross-dataset (deepfake) evaluation ----
    eval_celebdf_every_epoch = True
    celebdf_frames_per_video = 16
    celebdf_video_limit = 100           # subset size per eval pass (stratified by label); None = full ~518-video protocol list
    celebdf_crop_margin = 0.4

    num_workers = 8
    seed = 42
    device = "cuda" if torch.cuda.is_available() else "cpu"
    ckpt_dir = "/kaggle/working/ckpts"


def resolve_amp_dtype(cfg: "CFG"):
    if cfg.amp_dtype == "bf16" and torch.cuda.is_available() and torch.cuda.is_bf16_supported():
        return torch.bfloat16
    if cfg.amp_dtype == "bf16":
        print("[amp] bf16 requested but unsupported on this GPU -- falling back to fp16")
    return torch.float16


os.makedirs(CFG.ckpt_dir, exist_ok=True)
random.seed(CFG.seed); np.random.seed(CFG.seed); torch.manual_seed(CFG.seed)


# %% CELL 3 -- landmark graph utilities
def _ibug68_edges() -> List[Tuple[int, int]]:
    edges = []
    edges += [(i, i + 1) for i in range(0, 16)]                 # jaw
    edges += [(i, i + 1) for i in range(17, 21)]                # right eyebrow
    edges += [(i, i + 1) for i in range(22, 26)]                # left eyebrow
    edges += [(i, i + 1) for i in range(27, 30)]                # nose bridge
    edges += [(i, i + 1) for i in range(31, 35)]                # nose base
    edges += [(27, 31), (30, 33)]
    edges += [(i, i + 1) for i in range(36, 41)] + [(41, 36)]   # right eye loop
    edges += [(i, i + 1) for i in range(42, 47)] + [(47, 42)]   # left eye loop
    edges += [(i, i + 1) for i in range(48, 59)] + [(59, 48)]   # outer lip loop
    edges += [(i, i + 1) for i in range(60, 67)] + [(67, 60)]   # inner lip loop
    edges += [(19, 37), (24, 44), (33, 51)]                     # keep graph connected
    return edges


EDGES_68 = _ibug68_edges()
NUM_LANDMARKS = 68


def build_adjacency(num_nodes: int = NUM_LANDMARKS, edges=EDGES_68) -> np.ndarray:
    adj = np.zeros((num_nodes, num_nodes), dtype=bool)
    for i, j in edges:
        adj[i, j] = True
        adj[j, i] = True
    return adj


def build_hyper_adjacency(num_landmarks: int = NUM_LANDMARKS) -> np.ndarray:
    """(N+1)x(N+1) adjacency with a hyper-node at index 0, fully connected to
    every landmark (see deviation #3 at the top of this file)."""
    base = build_adjacency(num_landmarks)
    adj = np.zeros((num_landmarks + 1, num_landmarks + 1), dtype=bool)
    adj[1:, 1:] = base
    adj[0, 1:] = True
    adj[1:, 0] = True
    return adj


def all_pairs_hops(adj: np.ndarray, max_hop: int) -> np.ndarray:
    n = adj.shape[0]
    hop = np.full((n, n), -1, dtype=np.int64)
    for src in range(n):
        hop[src, src] = 0
        q = deque([src]); visited = {src}
        while q:
            u = q.popleft()
            if hop[src, u] >= max_hop:
                continue
            for v in np.nonzero(adj[u])[0]:
                if v not in visited:
                    visited.add(v)
                    hop[src, v] = hop[src, u] + 1
                    q.append(v)
    return hop


def node_degree(adj: np.ndarray) -> np.ndarray:
    return adj.sum(axis=1).astype(np.int64)


def landmark_to_patch_index(landmarks_xy: np.ndarray, img_size: int, patch_size: int) -> np.ndarray:
    grid = img_size // patch_size
    x = np.clip((landmarks_xy[:, 0] // patch_size).astype(np.int64), 0, grid - 1)
    y = np.clip((landmarks_xy[:, 1] // patch_size).astype(np.int64), 0, grid - 1)
    return y * grid + x


# %% CELL 4 -- landmark cache (one consolidated file, not one file per frame)
# Frames live under read-only /kaggle/input, so we can't write a sibling
# *_lmk.npy next to each image the way a locally-writable copy would allow.
# Everything is keyed by absolute frame path in a single dict, persisted to
# CFG.landmark_cache_path -- a few hundred bytes/frame, so even 100k frames
# is well under 100MB. Safe to re-run: already-cached paths are skipped, and
# progress is flushed periodically in case the kernel restarts mid-run.
class LandmarkCache:
    def __init__(self, cache_path, input_path=None):
        """Loads from `cache_path` (writable, /kaggle/working) if it already
        exists there (e.g. a resumed session); otherwise seeds from
        `input_path` if given -- a cache you published as a Kaggle Dataset
        on an earlier run, mounted read-only under /kaggle/input. Either
        way, `save()` always writes to `cache_path`; /kaggle/input can't be
        written to mid-session."""
        self.cache_path = cache_path
        if os.path.exists(cache_path):
            self.data = torch.load(cache_path, weights_only=False)
            print(f"[lmk-cache] loaded {len(self.data)} cached landmarks from {cache_path}")
        elif input_path and os.path.exists(input_path):
            self.data = torch.load(input_path, weights_only=False)
            print(f"[lmk-cache] seeded {len(self.data)} landmarks from published input cache {input_path}")
        else:
            self.data = {}

    def __contains__(self, key):
        return key in self.data

    def __len__(self):
        return len(self.data)

    def get(self, key):
        return self.data.get(key)

    def set(self, key, lmk):
        self.data[key] = lmk

    def save(self):
        os.makedirs(os.path.dirname(self.cache_path), exist_ok=True)
        torch.save(self.data, self.cache_path)


def build_landmark_cache(samples_by_split, cache_path, device="cuda", flush_every=2000, input_path=None):
    """samples_by_split: {split: [(video_id, frame_paths, label), ...]}.
    Populates/updates a LandmarkCache in place and returns it."""
    import face_alignment

    cache = LandmarkCache(cache_path, input_path=input_path)
    fa = face_alignment.FaceAlignment(
        face_alignment.LandmarksType.THREE_D, flip_input=False, device=device
    )
    n_ok, n_fail, n_skip, since_flush = 0, 0, 0, 0
    for split, samples in samples_by_split.items():
        for video_id, frame_paths, label in tqdm(samples, desc=f"landmarks[{split}]"):
            for fpath in frame_paths:
                if fpath in cache:
                    n_skip += 1
                    continue
                img = cv2.imread(fpath)
                if img is None:
                    cache.set(fpath, np.zeros((NUM_LANDMARKS, 3), dtype=np.float32))
                    n_fail += 1
                else:
                    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
                    preds = fa.get_landmarks(img_rgb)
                    if preds is None or len(preds) == 0:
                        lmk = np.zeros((NUM_LANDMARKS, 3), dtype=np.float32)
                        n_fail += 1
                    else:
                        lmk = preds[0].astype(np.float32)
                        if not np.isfinite(lmk).all():          # face-alignment can emit NaN/Inf
                            lmk = np.zeros((NUM_LANDMARKS, 3), dtype=np.float32)
                            n_fail += 1
                        else:
                            n_ok += 1
                    cache.set(fpath, lmk)
                since_flush += 1
                if since_flush >= flush_every:
                    cache.save()
                    since_flush = 0
    cache.save()
    print(f"[lmk-cache] {n_ok} ok, {n_fail} failed/no-face, {n_skip} already cached")
    return cache


# %% CELL 5 -- video dataset + augmentation
def joint_augment(imgs, lmks_px, img_size):
    """imgs: (T,3,H,W) float in [-1,1]; lmks_px: (T,68,2) pixel coords.
    Simplified augmentation (paper also mentions blur/sharpen/full affine)."""
    if random.random() < 0.5:
        imgs = torch.flip(imgs, dims=[-1])
        lmks_px = lmks_px.clone()
        lmks_px[..., 0] = img_size - 1 - lmks_px[..., 0]
    if random.random() < 0.5:
        angle = random.uniform(-10, 10)
        imgs = _rotate_batch(imgs, angle)
        lmks_px = _rotate_points(lmks_px, angle, img_size)
    return imgs, lmks_px


def _rotate_batch(imgs, angle_deg):
    theta = torch.tensor(
        [[math.cos(math.radians(angle_deg)), -math.sin(math.radians(angle_deg)), 0],
         [math.sin(math.radians(angle_deg)), math.cos(math.radians(angle_deg)), 0]],
        dtype=imgs.dtype,
    ).unsqueeze(0).repeat(imgs.shape[0], 1, 1)
    grid = F.affine_grid(theta, imgs.shape, align_corners=False)
    return F.grid_sample(imgs, grid, align_corners=False)


def _rotate_points(pts, angle_deg, img_size):
    c = img_size / 2.0
    a = math.radians(angle_deg)
    cos_a, sin_a = math.cos(a), math.sin(a)
    x = pts[..., 0] - c
    y = pts[..., 1] - c
    new_x = x * cos_a - y * sin_a + c
    new_y = x * sin_a + y * cos_a + c
    out = pts.clone()
    out[..., 0], out[..., 1] = new_x, new_y
    return out


class FASVideoDataset(Dataset):
    """samples: list of (video_id, [abs frame paths], label_str). Frames are
    read directly from wherever they already live (e.g. /kaggle/input) --
    nothing is copied (fix J). lmk_cache: a LandmarkCache (or plain dict)
    mapping frame path -> (68,3) landmark array; missing entries fall back
    to zeros."""
    LABEL2IDX = {"live": 1, "attack": 0}  # y=1 live, y=0 spoof (matches Eq.6)

    def __init__(self, samples, img_size: int = 224, num_frames: int = 8,
                 augment: bool = False, lmk_cache=None):
        self.samples = [(vid, paths, self.LABEL2IDX[label])
                         for vid, paths, label in samples if paths]
        if not self.samples:
            raise RuntimeError("FASVideoDataset got an empty sample list")
        self.img_size = img_size
        self.num_frames = num_frames
        self.augment = augment
        self.lmk_cache = lmk_cache

    def __len__(self):
        return len(self.samples)

    def _sample_frame_indices(self, n_available: int) -> list:
        T = self.num_frames
        if n_available >= T:
            return np.linspace(0, n_available - 1, T).round().astype(int).tolist()
        return list(range(n_available)) + [n_available - 1] * (T - n_available)

    def __getitem__(self, idx):
        video_id, frame_paths, label = self.samples[idx]
        chosen = self._sample_frame_indices(len(frame_paths))

        imgs, lmks = [], []
        for i in chosen:
            fpath = frame_paths[i]
            img = cv2.imread(fpath)
            h0, w0 = img.shape[:2]
            img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
            img = cv2.resize(img, (self.img_size, self.img_size))

            lmk = self.lmk_cache.get(fpath) if self.lmk_cache is not None else None
            if lmk is None or not np.isfinite(lmk).all():
                lmk = np.zeros((NUM_LANDMARKS, 3), dtype=np.float32)
            else:
                lmk = lmk.copy()
                sx, sy = self.img_size / w0, self.img_size / h0
                lmk[:, 0] *= sx
                lmk[:, 1] *= sy

            imgs.append(img)
            lmks.append(lmk)

        imgs = np.stack(imgs)
        lmks = np.stack(lmks).astype(np.float32)

        imgs = torch.from_numpy(imgs).permute(0, 3, 1, 2).float() / 255.0
        imgs = (imgs - 0.5) / 0.5
        lmks_px = torch.from_numpy(lmks[..., :2])
        lmks_z = torch.from_numpy(lmks[..., 2:3])

        if self.augment:
            imgs, lmks_px = joint_augment(imgs, lmks_px, self.img_size)

        lmks_norm = lmks_px.clone()
        lmks_norm[..., 0] /= self.img_size
        lmks_norm[..., 1] /= self.img_size
        z = lmks_z.squeeze(-1)
        z_range = (z.max() - z.min()) + 1e-6
        z_norm = (z - z.min()) / z_range
        lmks_norm = torch.cat([lmks_norm, z_norm.unsqueeze(-1)], dim=-1)

        return {
            "frames": imgs,                  # (T,3,H,W)
            "landmarks_px": lmks_px,          # (T,68,2) pixel coords, post-aug
            "landmarks_norm": lmks_norm,      # (T,68,3) in [0,1], post-aug
            "label": torch.tensor(label, dtype=torch.long),
        }


# %% CELL 6 -- Celeb-DF-v2 cross-dataset (deepfake) evaluation, no data copying
# Dataset: https://www.kaggle.com/datasets/reubensuju/celeb-df-v2
# Raw layout is Celeb-real/*.mp4, YouTube-real/*.mp4 (genuine) and
# Celeb-synthesis/*.mp4 (deepfake), plus an official List_of_testing_videos.txt
# protocol split. Unlike an earlier revision of this file, we do NOT extract
# and save cropped PNG frames -- `build_celebdf_index` walks each selected
# video once, running face detection only to record a crop box + landmarks
# per sampled frame (a few hundred bytes), never the pixels. At training
# time `CelebDFVideoDataset` re-opens the source .mp4 straight from
# /kaggle/input and decodes just the frames it needs.

# EDIT to match wherever Kaggle mounted your attached celeb-df-v2 dataset.
CELEBDF_INPUT_ROOT = "/kaggle/input/datasets/reubensuju/celeb-df-v2"


def _find_celebdf_root():
    if not os.path.isdir(CELEBDF_INPUT_ROOT):
        raise FileNotFoundError(
            f"CELEBDF_INPUT_ROOT={CELEBDF_INPUT_ROOT} does not exist. Attach the "
            f"reubensuju/celeb-df-v2 dataset and set CELEBDF_INPUT_ROOT to its mounted path."
        )
    return CELEBDF_INPUT_ROOT


def _find_file(root, filename):
    for dirpath, _, filenames in os.walk(root):
        if filename in filenames:
            return os.path.join(dirpath, filename)
    return None


def load_celebdf_protocol(celebdf_root):
    """Returns a list of (relative_video_path, 'live'|'attack')."""
    list_path = _find_file(celebdf_root, "List_of_testing_videos.txt")
    items = []
    if list_path is not None:
        with open(list_path) as f:
            for line in f:
                parts = line.strip().split()
                if len(parts) != 2:
                    continue
                _, rel_path = parts
                rel_path = rel_path.replace("\\", "/")
                # Trust the folder name, not the file's numeric label: mirrors of
                # List_of_testing_videos.txt disagree on 0/1 polarity, but
                # "Celeb-synthesis" vs "Celeb-real"/"YouTube-real" is unambiguous.
                label = "attack" if "synthesis" in rel_path.lower() else "live"
                items.append((rel_path, label))
        print(f"[celebdf] official protocol list: {list_path} ({len(items)} videos)")
    else:
        print("[celebdf] List_of_testing_videos.txt not found -- using ALL videos "
              "(slower; set CFG.celebdf_video_limit to cap it).")
        for folder, label in [("Celeb-real", "live"), ("YouTube-real", "live"),
                               ("Celeb-synthesis", "attack")]:
            folder_path = os.path.join(celebdf_root, folder)
            if not os.path.isdir(folder_path):
                continue
            for fname in sorted(os.listdir(folder_path)):
                if fname.lower().endswith(".mp4"):
                    items.append((f"{folder}/{fname}", label))
    return items


def _stratified_subset(items, limit, seed=42):
    """Picks `limit` items from (path,label) pairs, split proportionally
    across labels and shuffled with a fixed seed. Plain items[:limit] risks
    a skewed subset (e.g. mostly one class) if the protocol file happens to
    group videos by label, which would give a misleading HTER/AUC."""
    by_label = defaultdict(list)
    for item in items:
        by_label[item[1]].append(item)

    rng = random.Random(seed)
    for label in by_label:
        rng.shuffle(by_label[label])

    total = len(items)
    subset = []
    for label, group in by_label.items():
        n = max(1, round(limit * len(group) / total)) if total else 0
        subset.extend(group[:n])
    rng.shuffle(subset)
    return subset[:limit]


def _crop_box_and_landmarks(preds, h, w, margin, img_size):
    """Returns ((x0,y0,x1,y1) in the ORIGINAL frame, landmarks already
    transformed into the FINAL img_size x img_size crop's pixel space --
    same convention FASVideoDataset uses, so downstream code doesn't need
    to know the crop box at all after this point)."""
    if not preds:
        # No face found -- generous center crop rather than dropping the frame
        # (Celeb-DF clips are talking-head videos, face is almost always centered).
        side = int(min(h, w) * 0.8)
        y0, x0 = (h - side) // 2, (w - side) // 2
        x1, y1 = x0 + side, y0 + side
        lmk = np.zeros((NUM_LANDMARKS, 3), dtype=np.float32)
        return (x0, y0, x1, y1), lmk

    p = preds[0].astype(np.float32)  # (68,3)
    xy = p[:, :2]
    x_min, y_min = xy.min(axis=0)
    x_max, y_max = xy.max(axis=0)
    bw, bh = max(x_max - x_min, 1.0), max(y_max - y_min, 1.0)
    cx, cy = (x_min + x_max) / 2, (y_min + y_max) / 2
    side = max(bw, bh) * (1.0 + margin)
    x0, x1 = int(max(0, cx - side / 2)), int(min(w, cx + side / 2))
    y0, y1 = int(max(0, cy - side / 2)), int(min(h, cy + side / 2))
    if x1 <= x0 or y1 <= y0:
        x0, y0, x1, y1 = 0, 0, w, h

    sx, sy = img_size / (x1 - x0), img_size / (y1 - y0)
    lmk = p.copy()
    lmk[:, 0] = (lmk[:, 0] - x0) * sx
    lmk[:, 1] = (lmk[:, 1] - y0) * sy
    if not np.isfinite(lmk).all():
        lmk = np.zeros((NUM_LANDMARKS, 3), dtype=np.float32)
    return (x0, y0, x1, y1), lmk


def build_celebdf_index(celebdf_root, fa_model, num_frames_to_cache=16,
                         img_size=224, crop_margin=0.4, limit=None):
    """Returns a list of {"video_path", "label" (0/1), "frames": [
    {"frame_idx", "box", "lmk_crop"}, ...]} -- metadata only, no image files
    written anywhere."""
    items = load_celebdf_protocol(celebdf_root)
    if limit:
        items = _stratified_subset(items, limit)
    index = []
    for rel_path, label in tqdm(items, desc="celebdf: indexing (metadata only, no frames written)"):
        video_path = os.path.join(celebdf_root, rel_path)
        if not os.path.isfile(video_path):
            continue
        cap = cv2.VideoCapture(video_path)
        total = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
        if total <= 0:
            cap.release()
            continue
        target_set = set(np.linspace(0, total - 1, min(num_frames_to_cache, total)).round().astype(int).tolist())

        frame_meta, i = [], 0
        while len(frame_meta) < len(target_set):
            ok = cap.grab()
            if not ok:
                break
            if i in target_set:
                ok, frame = cap.retrieve()
                if ok:
                    h, w = frame.shape[:2]
                    frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
                    try:
                        preds = fa_model.get_landmarks(frame_rgb)
                    except Exception:
                        preds = None
                    box, lmk = _crop_box_and_landmarks(preds, h, w, crop_margin, img_size)
                    frame_meta.append({"frame_idx": i, "box": box, "lmk_crop": lmk})
            i += 1
        cap.release()
        if frame_meta:
            index.append({
                "video_path": video_path,
                "label": 1 if label == "live" else 0,
                "frames": frame_meta,
            })
    print(f"[celebdf] indexed {len(index)}/{len(items)} videos (metadata-only, no frame files written)")
    return index


def build_celebdf_index_cached(cfg: "CFG"):
    if os.path.exists(cfg.celebdf_index_path):
        index = torch.load(cfg.celebdf_index_path, weights_only=False)  # our own cache, not a state_dict
        print(f"[celebdf] loaded cached index: {len(index)} videos from {cfg.celebdf_index_path}")
        return index
    if cfg.celebdf_index_input_path and os.path.exists(cfg.celebdf_index_input_path):
        index = torch.load(cfg.celebdf_index_input_path, weights_only=False)
        print(f"[celebdf] seeded from published input index: {len(index)} videos "
              f"from {cfg.celebdf_index_input_path}")
        os.makedirs(os.path.dirname(cfg.celebdf_index_path), exist_ok=True)
        torch.save(index, cfg.celebdf_index_path)  # mirror into /kaggle/working for later cells
        return index

    import face_alignment
    celebdf_root = _find_celebdf_root()
    fa_model = face_alignment.FaceAlignment(
        face_alignment.LandmarksType.THREE_D, flip_input=False, device=cfg.device
    )
    index = build_celebdf_index(
        celebdf_root, fa_model,
        num_frames_to_cache=cfg.celebdf_frames_per_video,
        img_size=cfg.img_size, crop_margin=cfg.celebdf_crop_margin,
        limit=cfg.celebdf_video_limit,
    )
    del fa_model
    if cfg.device == "cuda":
        torch.cuda.empty_cache()

    os.makedirs(os.path.dirname(cfg.celebdf_index_path), exist_ok=True)
    torch.save(index, cfg.celebdf_index_path)
    return index


class CelebDFVideoDataset(Dataset):
    """Reads frames directly from the original .mp4 files under
    /kaggle/input at __getitem__ time using the crop box + landmarks
    precomputed once by `build_celebdf_index` -- no cropped frames are ever
    written to disk (fix J)."""

    def __init__(self, index, img_size: int = 224, num_frames: int = 8):
        self.index = index
        self.img_size = img_size
        self.num_frames = num_frames

    def __len__(self):
        return len(self.index)

    def _sample_frame_indices(self, n_available: int) -> list:
        T = self.num_frames
        if n_available >= T:
            return np.linspace(0, n_available - 1, T).round().astype(int).tolist()
        return list(range(n_available)) + [n_available - 1] * (T - n_available)

    def __getitem__(self, idx):
        entry = self.index[idx]
        video_path = entry["video_path"]
        frame_meta = entry["frames"]
        chosen_meta = [frame_meta[i] for i in self._sample_frame_indices(len(frame_meta))]
        target_idx = {m["frame_idx"] for m in chosen_meta}

        raw_frames = {}
        cap = cv2.VideoCapture(video_path)
        i = 0
        while len(raw_frames) < len(target_idx):
            ok = cap.grab()
            if not ok:
                break
            if i in target_idx:
                ok, frame = cap.retrieve()
                if ok:
                    raw_frames[i] = frame
            i += 1
        cap.release()

        imgs, lmks = [], []
        for m in chosen_meta:
            frame = raw_frames.get(m["frame_idx"])
            if frame is None:
                crop = np.zeros((self.img_size, self.img_size, 3), dtype=np.uint8)
            else:
                x0, y0, x1, y1 = m["box"]
                crop = frame[y0:y1, x0:x1]
                if crop.size == 0:
                    crop = frame
                crop = cv2.cvtColor(crop, cv2.COLOR_BGR2RGB)
                crop = cv2.resize(crop, (self.img_size, self.img_size))
            imgs.append(crop)
            lmks.append(m["lmk_crop"])

        imgs = np.stack(imgs)
        lmks = np.stack(lmks).astype(np.float32)

        imgs = torch.from_numpy(imgs).permute(0, 3, 1, 2).float() / 255.0
        imgs = (imgs - 0.5) / 0.5
        lmks_px = torch.from_numpy(lmks[..., :2])
        lmks_z = torch.from_numpy(lmks[..., 2:3])

        lmks_norm = lmks_px.clone()
        lmks_norm[..., 0] /= self.img_size
        lmks_norm[..., 1] /= self.img_size
        z = lmks_z.squeeze(-1)
        z_range = (z.max() - z.min()) + 1e-6
        z_norm = (z - z.min()) / z_range
        lmks_norm = torch.cat([lmks_norm, z_norm.unsqueeze(-1)], dim=-1)

        return {
            "frames": imgs,
            "landmarks_px": lmks_px,
            "landmarks_norm": lmks_norm,
            "label": torch.tensor(entry["label"], dtype=torch.long),
        }


def build_celebdf_eval_loader(cfg: "CFG"):
    index = build_celebdf_index_cached(cfg)
    ds = CelebDFVideoDataset(index, img_size=cfg.img_size, num_frames=cfg.num_frames)
    loader = DataLoader(ds, batch_size=cfg.finetune_batch_size, shuffle=False,
                         num_workers=cfg.num_workers, pin_memory=True)
    return loader


# %% CELL 7 -- shared embedding modules
class PatchEmbed(nn.Module):
    def __init__(self, img_size=224, patch_size=16, in_chans=3, embed_dim=384):
        super().__init__()
        self.grid = img_size // patch_size
        self.num_patches = self.grid * self.grid
        self.patch_size = patch_size
        self.proj = nn.Conv2d(in_chans, embed_dim, kernel_size=patch_size, stride=patch_size)

    def forward(self, x):
        B, T, C, H, W = x.shape
        x = x.reshape(B * T, C, H, W)
        x = self.proj(x).flatten(2).transpose(1, 2)  # (B*T, L, D)
        D = x.shape[-1]
        return x.reshape(B, T, -1, D)


def sine_cosine_pe(coords, num_freq_bands):
    """Eq. 1. coords: (...,C) in roughly [0,1] -> (...,C*2*F)."""
    freqs = (2.0 ** torch.arange(num_freq_bands, device=coords.device, dtype=coords.dtype)) * math.pi
    args = coords.unsqueeze(-1) * freqs
    enc = torch.cat([torch.sin(args), torch.cos(args)], dim=-1)
    return enc.flatten(-2)


class NodeEmbed(nn.Module):
    """Eq. 9: h_vi = NodeEmbed(gamma(x_vi)) + DegreeEmbed(vi)."""

    def __init__(self, embed_dim, num_freq_bands, num_nodes=NUM_LANDMARKS, max_degree=8):
        super().__init__()
        in_dim = 3 * 2 * num_freq_bands
        self.proj = nn.Linear(in_dim, embed_dim)
        degree = torch.from_numpy(node_degree(build_adjacency(num_nodes))).long()
        self.register_buffer("degree", degree)
        self.degree_embed = nn.Embedding(max_degree + 1, embed_dim)
        self.num_freq_bands = num_freq_bands

    def forward(self, landmarks_norm):
        pe = sine_cosine_pe(landmarks_norm, self.num_freq_bands)
        h = self.proj(pe)
        deg = self.degree.clamp(max=self.degree_embed.num_embeddings - 1)
        return h + self.degree_embed(deg)[None, None, :, :]


class MaskToken(nn.Module):
    def __init__(self, dim):
        super().__init__()
        self.token = nn.Parameter(torch.zeros(1, 1, 1, dim))
        nn.init.trunc_normal_(self.token, std=0.02)

    def forward(self, x, mask):
        # x: (B,T,L,D), mask: (B,T,L) bool, True = replace with mask token
        tok = self.token.expand_as(x)
        return torch.where(mask.unsqueeze(-1), tok, x)


# %% CELL 8 -- spatial attention (vision stream), Eq. 2-5
class RelPosBias(nn.Module):
    """Learnable relative-position bias over the LxL patch grid. `has_cls`
    zero-pads to (L+1)x(L+1) at forward time if a class token is prepended
    (padding doesn't touch stored parameters, so pretrain<->finetune weight
    transfer is unaffected by this flag)."""

    def __init__(self, grid, num_heads):
        super().__init__()
        self.grid = grid
        num_rel = (2 * grid - 1) * (2 * grid - 1)
        self.table = nn.Parameter(torch.zeros(num_rel, num_heads))
        coords = torch.stack(torch.meshgrid(torch.arange(grid), torch.arange(grid), indexing="ij"), dim=-1).reshape(-1, 2)
        rel = coords[:, None, :] - coords[None, :, :] + (grid - 1)
        idx = rel[..., 0] * (2 * grid - 1) + rel[..., 1]
        self.register_buffer("index", idx)
        nn.init.trunc_normal_(self.table, std=0.02)

    def forward(self, has_cls: bool):
        L = self.grid * self.grid
        bias = self.table[self.index.reshape(-1)].reshape(L, L, -1).permute(2, 0, 1)  # (H,L,L)
        if has_cls:
            H = bias.shape[0]
            full = bias.new_zeros(H, L + 1, L + 1)
            full[:, 1:, 1:] = bias
            bias = full
        return bias


class SpatialAttention(nn.Module):
    """Applied independently per frame across spatial tokens (Eq. 2-5)."""

    def __init__(self, dim, num_heads, grid, has_cls=True):
        super().__init__()
        self.num_heads = num_heads
        self.head_dim = dim // num_heads
        self.scale = self.head_dim ** -0.5
        self.qkv = nn.Linear(dim, dim * 3)
        self.proj = nn.Linear(dim, dim)
        self.rel_pos = RelPosBias(grid, num_heads)
        self.has_cls = has_cls

    def forward(self, x):
        # x: (B,T,L,D)  (L includes the class token slot iff self.has_cls)
        B, T, L, D = x.shape
        qkv = self.qkv(x).reshape(B, T, L, 3, self.num_heads, self.head_dim).permute(3, 0, 1, 4, 2, 5)
        q, k, v = qkv[0], qkv[1], qkv[2]
        attn = (q @ k.transpose(-2, -1)) * self.scale  # (B,T,H,L,L)
        attn = attn + self.rel_pos(self.has_cls)[None, None]
        attn = attn.softmax(dim=-1)
        out = attn @ v                                  # (B,T,H,L,hd)
        out = out.transpose(2, 3).reshape(B, T, L, D)    # merge heads -> D
        return self.proj(out), attn


# %% CELL 9 -- topology-aware spatial attention (graph stream), Eq. 7-9
class TopoSpatialAttention(nn.Module):
    def __init__(self, dim, num_heads, num_landmarks=NUM_LANDMARKS, max_hop=2):
        super().__init__()
        self.num_heads = num_heads
        self.head_dim = dim // num_heads
        self.scale = self.head_dim ** -0.5
        self.qkv = nn.Linear(dim, dim * 3)
        self.proj = nn.Linear(dim, dim)

        adj = build_hyper_adjacency(num_landmarks)   # (N+1,N+1), hyper-node at index 0
        hop = all_pairs_hops(adj, max_hop)
        self.register_buffer("hop", torch.from_numpy(hop).long())
        self.hop_embed = nn.Parameter(torch.zeros(max_hop + 1, num_heads))
        nn.init.trunc_normal_(self.hop_embed, std=0.02)

    def _bias(self, device):
        N = self.hop.shape[0]
        valid = self.hop >= 0
        idx = self.hop.clamp(min=0)
        gathered = self.hop_embed[idx.reshape(-1)].reshape(N, N, self.num_heads)
        bias = torch.where(valid.unsqueeze(-1), gathered, torch.full_like(gathered, NEG_INF))
        return bias.permute(2, 0, 1)  # (H,N,N)

    def forward(self, x):
        # x: (B,T,N+1,D) -- index 0 is always the hyper-node
        B, T, N, D = x.shape
        qkv = self.qkv(x).reshape(B, T, N, 3, self.num_heads, self.head_dim).permute(3, 0, 1, 4, 2, 5)
        q, k, v = qkv[0], qkv[1], qkv[2]
        attn = (q @ k.transpose(-2, -1)) * self.scale
        attn = attn + self._bias(x.device)[None, None]
        attn = attn.softmax(dim=-1)
        out = attn @ v
        out = out.transpose(2, 3).reshape(B, T, N, D)
        return self.proj(out), attn


# %% CELL 10 -- Kronecker Temporal Attention (KTA), Eq. 10-14, shared by both streams
@functools.lru_cache(maxsize=8)
def kronecker_temporal_mask(T, L, device, dtype):
    I_t = torch.eye(T, device=device)
    J_m = torch.ones(L, L, device=device)
    I_m = torch.eye(L, device=device)
    M = torch.kron(I_t, (J_m - I_m))  # ==1 at same-timestep,different-patch entries
    return torch.where(M > 0, torch.full_like(M, NEG_INF), torch.zeros_like(M)).to(dtype)


class KroneckerTemporalAttention(nn.Module):
    def __init__(self, dim, num_heads):
        super().__init__()
        self.num_heads = num_heads
        self.head_dim = dim // num_heads
        self.scale = self.head_dim ** -0.5
        self.qkv = nn.Linear(dim, dim * 3)
        self.proj = nn.Linear(dim, dim)

    def forward(self, x, guide_bias=None):
        """x: (B,T,L,D). guide_bias: optional (B,T*L,T*L) additive bias
        (already scattered from the graph stream, Eq.16), broadcast over heads.
        Returns out (B,T,L,D) and raw pre-softmax logits (B,H,T*L,T*L)."""
        B, T, L, D = x.shape
        x_flat = x.reshape(B, T * L, D)
        qkv = self.qkv(x_flat).reshape(B, T * L, 3, self.num_heads, self.head_dim).permute(2, 0, 3, 1, 4)
        q, k, v = qkv[0], qkv[1], qkv[2]
        raw_logits = (q @ k.transpose(-2, -1)) * self.scale

        mask = kronecker_temporal_mask(T, L, x.device, raw_logits.dtype)[None, None]
        logits = raw_logits + mask
        if guide_bias is not None:
            logits = logits + guide_bias[:, None]

        attn = logits.softmax(dim=-1)
        out = (attn @ v).transpose(1, 2).reshape(B, T * L, D)
        out = self.proj(out).reshape(B, T, L, D)
        return out, raw_logits


def extract_landmark_block(raw_logits, T, N):
    """raw_logits: (B,H,T*(N+1),T*(N+1)) from the graph stream's KTA (tokens
    are flattened as t=0:[hyper,n1..nN], t=1:[hyper,n1..nN],...). Returns the
    (B,H,T*N,T*N) sub-block with the hyper-node slot dropped at EVERY t
    (a naive [:,:,1:,1:] slice only drops it once, at t=0)."""
    B, H = raw_logits.shape[0], raw_logits.shape[1]
    Np1 = N + 1
    r = raw_logits.view(B, H, T, Np1, T, Np1)
    r = r[:, :, :, 1:, :, 1:]
    return r.reshape(B, H, T * N, T * N)


def scatter_graph_bias_to_vision(graph_raw_logits_land, patch_idx, T, N, L, bias_clip=8.0):
    """graph_raw_logits_land: (B,H,T*N,T*N) landmark-only sub-block (already
    hyper-node-stripped via extract_landmark_block). patch_idx: (B,T,N) long,
    landmark -> vision patch index in [0,L). Returns (B,T*L,T*L) additive bias.

    FIX (see docstring item A): multiple landmarks routinely collide into the
    same vision patch (68 landmarks onto a 14x14 grid). The original code
    used scatter_ADD, so the injected bias grew unboundedly with collision
    count instead of the natural attention-logit scale -- the actual cause of
    the NaN-loss bug. This computes a scatter-MEAN (divide by contribution
    count) and clamps to +-bias_clip, keeping the graph-guidance signal
    informative but numerically bounded regardless of landmark density.
    Accumulation happens in fp32 for numerical stability under autocast, and
    the whole thing is vectorized over the batch (was a Python for-loop).
    """
    B = graph_raw_logits_land.shape[0]
    device = graph_raw_logits_land.device
    out_dtype = graph_raw_logits_land.dtype
    graph_bias = graph_raw_logits_land.float().mean(dim=1)  # (B,T*N,T*N), average heads (deviation #6)

    TL, TN = T * L, T * N
    t_idx = torch.arange(T, device=device).repeat_interleave(N)          # (TN,) same for every batch item
    node_to_patch = patch_idx.reshape(B, TN)                             # (B,TN) landmark -> vision patch
    row_full = t_idx.unsqueeze(0) * L + node_to_patch                    # (B,TN) -> flat position in [0, T*L)

    idx_i = row_full.unsqueeze(2).expand(B, TN, TN).reshape(B, -1)
    idx_j = row_full.unsqueeze(1).expand(B, TN, TN).reshape(B, -1)
    batch_offset = (torch.arange(B, device=device) * TL * TL).unsqueeze(1)
    flat_idx = (idx_i * TL + idx_j + batch_offset).reshape(-1)

    values = graph_bias.reshape(-1)
    out_sum = torch.zeros(B * TL * TL, device=device, dtype=torch.float32)
    out_cnt = torch.zeros(B * TL * TL, device=device, dtype=torch.float32)
    out_sum.scatter_add_(0, flat_idx, values)
    out_cnt.scatter_add_(0, flat_idx, torch.ones_like(values))

    out = (out_sum / out_cnt.clamp(min=1.0)).reshape(B, TL, TL)
    out = out.clamp(min=-bias_clip, max=bias_clip)
    return out.to(out_dtype)


# %% CELL 11 -- spatiotemporal blocks
class Mlp(nn.Module):
    def __init__(self, dim, ratio=4.0, drop=0.0):
        super().__init__()
        hidden = int(dim * ratio)
        self.fc1 = nn.Linear(dim, hidden)
        self.act = nn.GELU()
        self.fc2 = nn.Linear(hidden, dim)
        self.drop = nn.Dropout(drop)

    def forward(self, x):
        return self.fc2(self.drop(self.act(self.fc1(x))))


class VisionSTB(nn.Module):
    def __init__(self, dim, num_heads, grid, mlp_ratio=4.0, has_cls=True):
        super().__init__()
        self.norm1 = nn.LayerNorm(dim)
        self.spatial = SpatialAttention(dim, num_heads, grid, has_cls=has_cls)
        self.norm2 = nn.LayerNorm(dim)
        self.mlp1 = Mlp(dim, mlp_ratio)
        self.norm3 = nn.LayerNorm(dim)
        self.temporal = KroneckerTemporalAttention(dim, num_heads)
        self.norm4 = nn.LayerNorm(dim)
        self.mlp2 = Mlp(dim, mlp_ratio)

    def forward(self, x, guide_bias=None):
        sa_out, _ = self.spatial(self.norm1(x))
        x = x + sa_out
        x = x + self.mlp1(self.norm2(x))
        ta_out, raw_logits = self.temporal(self.norm3(x), guide_bias=guide_bias)
        x = x + ta_out
        x = x + self.mlp2(self.norm4(x))
        return x, raw_logits


class GraphSTB(nn.Module):
    def __init__(self, dim, num_heads, num_landmarks, max_hop, mlp_ratio=4.0):
        super().__init__()
        self.norm1 = nn.LayerNorm(dim)
        self.spatial = TopoSpatialAttention(dim, num_heads, num_landmarks, max_hop)
        self.norm2 = nn.LayerNorm(dim)
        self.mlp1 = Mlp(dim, mlp_ratio)
        self.norm3 = nn.LayerNorm(dim)
        self.temporal = KroneckerTemporalAttention(dim, num_heads)
        self.norm4 = nn.LayerNorm(dim)
        self.mlp2 = Mlp(dim, mlp_ratio)

    def forward(self, x):
        sa_out, _ = self.spatial(self.norm1(x))
        x = x + sa_out
        x = x + self.mlp1(self.norm2(x))
        ta_out, raw_logits = self.temporal(self.norm3(x))
        x = x + ta_out
        x = x + self.mlp2(self.norm4(x))
        return x, raw_logits


# %% CELL 12 -- full fine-tuning model
class G2V2former(nn.Module):
    def __init__(self, cfg: CFG):
        super().__init__()
        self.T = cfg.num_frames
        self.N = cfg.num_landmarks
        self.patch_size = cfg.patch_size
        self.img_size = cfg.img_size
        self.guide_bias_clip = cfg.guide_bias_clip

        # vision stream
        self.patch_embed = PatchEmbed(cfg.img_size, cfg.patch_size, 3, cfg.vis_dim)
        self.grid = self.patch_embed.grid
        self.L = self.patch_embed.num_patches
        self.vis_pos_embed = nn.Parameter(torch.zeros(1, 1, self.L, cfg.vis_dim))
        self.vis_time_embed = nn.Parameter(torch.zeros(1, cfg.num_frames, 1, cfg.vis_dim))
        self.vis_cls_token = nn.Parameter(torch.zeros(1, 1, 1, cfg.vis_dim))
        self.vis_blocks = nn.ModuleList([
            VisionSTB(cfg.vis_dim, cfg.vis_heads, self.grid, cfg.mlp_ratio, has_cls=True)
            for _ in range(cfg.vis_depth)
        ])
        self.vis_norm = nn.LayerNorm(cfg.vis_dim)

        # graph stream
        self.node_embed = NodeEmbed(cfg.graph_dim, cfg.num_freq_bands, cfg.num_landmarks)
        self.hyper_node = nn.Parameter(torch.zeros(1, 1, 1, cfg.graph_dim))
        self.graph_time_embed = nn.Parameter(torch.zeros(1, cfg.num_frames, 1, cfg.graph_dim))
        self.graph_blocks = nn.ModuleList([
            GraphSTB(cfg.graph_dim, cfg.graph_heads, cfg.num_landmarks, cfg.max_hop, cfg.mlp_ratio)
            for _ in range(cfg.graph_depth)
        ])
        self.graph_norm = nn.LayerNorm(cfg.graph_dim)

        self.head = nn.Linear(cfg.vis_dim + cfg.graph_dim, cfg.num_classes)

        nn.init.trunc_normal_(self.vis_pos_embed, std=0.02)
        nn.init.trunc_normal_(self.vis_time_embed, std=0.02)
        nn.init.trunc_normal_(self.vis_cls_token, std=0.02)
        nn.init.trunc_normal_(self.hyper_node, std=0.02)
        nn.init.trunc_normal_(self.graph_time_embed, std=0.02)

    def forward(self, frames, landmarks_norm, landmarks_px):
        B, T = frames.shape[0], frames.shape[1]

        x = self.patch_embed(frames)
        x = x + self.vis_pos_embed + self.vis_time_embed
        cls_tok = self.vis_cls_token.expand(B, T, -1, -1)
        x = torch.cat([cls_tok, x], dim=2)     # (B,T,1+L,D)
        L_full = x.shape[2]

        g = self.node_embed(landmarks_norm) + self.graph_time_embed
        hnode = self.hyper_node.expand(B, T, -1, -1)
        g = torch.cat([hnode, g], dim=2)       # (B,T,1+N,D)

        with torch.no_grad():
            grid = self.grid
            px = landmarks_px.clamp(min=0, max=self.img_size - 1)
            col = (px[..., 0] // self.patch_size).long().clamp(max=grid - 1)
            row = (px[..., 1] // self.patch_size).long().clamp(max=grid - 1)
            patch_idx = (row * grid + col) + 1  # +1 for the prepended vision cls slot

        guide_bias = None
        for vblock, gblock in zip(self.vis_blocks, self.graph_blocks):
            g, g_raw_logits = gblock(g)
            g_land_logits = extract_landmark_block(g_raw_logits, T, self.N)
            guide_bias = scatter_graph_bias_to_vision(
                g_land_logits, patch_idx, T, self.N, L_full, bias_clip=self.guide_bias_clip
            )
            x, _ = vblock(x, guide_bias=guide_bias)

        x = self.vis_norm(x)
        g = self.graph_norm(g)

        vis_cls_seq = x[:, :, 0, :]              # (B,T,D) per-frame class token, for PCL
        vis_cls_out = vis_cls_seq.mean(dim=1)     # (B,D) for classification
        graph_hyper_out = g[:, :, 0, :].mean(dim=1)

        feat = torch.cat([vis_cls_out, graph_hyper_out], dim=-1)
        logits = self.head(feat)
        return logits, vis_cls_out, graph_hyper_out, vis_cls_seq


# %% CELL 13 -- Photometric Consistency Loss (PCL), Eq. 6
def photometric_consistency_loss(cls_seq, labels, tau=0.07):
    """cls_seq: (B,T,D) per-frame vision class tokens. labels: (B,) 1=live,0=spoof.
    Live frames are pulled together across ALL videos; spoof frames are only
    pulled together within the SAME video (Eq. 6)."""
    B, T, D = cls_seq.shape
    device = cls_seq.device
    z = F.normalize(cls_seq.float().reshape(B * T, D), dim=-1)  # fp32 for stable exp/log under autocast
    sim = (z @ z.t()) / tau
    BT = B * T
    y = labels.repeat_interleave(T)
    vid = torch.arange(B, device=device).repeat_interleave(T)
    not_self = ~torch.eye(BT, dtype=torch.bool, device=device)

    exp_sim = torch.exp(sim) * not_self
    denom = exp_sim.sum(dim=1).clamp(min=1e-8)
    log_prob = sim - torch.log(denom).unsqueeze(1)

    losses = []
    for i in range(BT):
        same_label = (y == y[i]) & not_self[i]
        if y[i].item() == 1:
            pos = same_label
        else:
            pos = same_label & (vid == vid[i])
        if pos.sum() == 0:
            continue
        losses.append(-log_prob[i, pos].mean())
    if not losses:
        return cls_seq.new_zeros(())
    return torch.stack(losses).mean()


# %% CELL 14 -- MAE pretraining models (SimMIM-style, see deviation #5)
class VisionMAE(nn.Module):
    def __init__(self, cfg: CFG):
        super().__init__()
        self.cfg = cfg
        self.patch_embed = PatchEmbed(cfg.img_size, cfg.patch_size, 3, cfg.vis_dim)
        self.grid = self.patch_embed.grid
        self.L = self.patch_embed.num_patches
        self.vis_pos_embed = nn.Parameter(torch.zeros(1, 1, self.L, cfg.vis_dim))
        self.vis_time_embed = nn.Parameter(torch.zeros(1, cfg.num_frames, 1, cfg.vis_dim))
        self.mask_token = MaskToken(cfg.vis_dim)
        self.vis_blocks = nn.ModuleList([
            VisionSTB(cfg.vis_dim, cfg.vis_heads, self.grid, cfg.mlp_ratio, has_cls=False)
            for _ in range(cfg.vis_depth)
        ])
        self.enc_norm = nn.LayerNorm(cfg.vis_dim)

        dd = cfg.decoder_dim_vis
        dh = max(1, dd // 64)
        self.decoder_embed = nn.Linear(cfg.vis_dim, dd)
        self.decoder_pos = nn.Parameter(torch.zeros(1, 1, self.L, dd))
        self.decoder_time = nn.Parameter(torch.zeros(1, cfg.num_frames, 1, dd))
        self.decoder_blocks = nn.ModuleList([
            VisionSTB(dd, dh, self.grid, cfg.mlp_ratio, has_cls=False)
            for _ in range(cfg.decoder_depth)
        ])
        self.decoder_norm = nn.LayerNorm(dd)
        self.decoder_pred = nn.Linear(dd, cfg.patch_size * cfg.patch_size * 3)

        for p in (self.vis_pos_embed, self.vis_time_embed, self.decoder_pos, self.decoder_time):
            nn.init.trunc_normal_(p, std=0.02)

    def patchify(self, imgs):
        B, T, C, H, W = imgs.shape
        p = self.patch_embed.patch_size
        x = imgs.reshape(B * T, C, H, W).unfold(2, p, p).unfold(3, p, p)  # (BT,C,gh,gw,p,p)
        x = x.permute(0, 2, 3, 1, 4, 5).reshape(B * T, self.grid * self.grid, C * p * p)
        return x.reshape(B, T, self.grid * self.grid, C * p * p)

    def random_mask(self, B, T, device, ratio):
        """Vectorized: exactly floor(L*ratio) True positions per (b,t) row,
        picked uniformly at random (was a Python double for-loop calling
        torch.randperm once per (b,t))."""
        L = self.L
        n_mask = int(L * ratio)
        noise = torch.rand(B, T, L, device=device)
        thresh = noise.kthvalue(n_mask, dim=-1, keepdim=True).values
        return noise <= thresh

    def forward(self, frames, ratio):
        B, T = frames.shape[0], frames.shape[1]
        target = self.patchify(frames)
        x = self.patch_embed(frames) + self.vis_pos_embed + self.vis_time_embed
        mask = self.random_mask(B, T, frames.device, ratio)
        x = self.mask_token(x, mask)
        for blk in self.vis_blocks:
            x, _ = blk(x)
        x = self.enc_norm(x)

        d = self.decoder_embed(x) + self.decoder_pos + self.decoder_time
        for blk in self.decoder_blocks:
            d, _ = blk(d)
        d = self.decoder_norm(d)
        pred = self.decoder_pred(d)  # (B,T,L,patch*patch*3)

        loss_map = ((pred - target) ** 2).mean(dim=-1)  # (B,T,L)
        loss = (loss_map * mask.float()).sum() / mask.float().sum().clamp(min=1)
        return loss


class GraphMAE(nn.Module):
    def __init__(self, cfg: CFG):
        super().__init__()
        self.cfg = cfg
        self.node_embed = NodeEmbed(cfg.graph_dim, cfg.num_freq_bands, cfg.num_landmarks)
        self.graph_time_embed = nn.Parameter(torch.zeros(1, cfg.num_frames, 1, cfg.graph_dim))
        self.hyper_token = nn.Parameter(torch.zeros(1, 1, 1, cfg.graph_dim))  # never masked
        self.mask_token = MaskToken(cfg.graph_dim)
        self.graph_blocks = nn.ModuleList([
            GraphSTB(cfg.graph_dim, cfg.graph_heads, cfg.num_landmarks, cfg.max_hop, cfg.mlp_ratio)
            for _ in range(cfg.graph_depth)
        ])
        self.enc_norm = nn.LayerNorm(cfg.graph_dim)

        dd = cfg.decoder_dim_graph
        dh = max(1, dd // 24)
        self.decoder_embed = nn.Linear(cfg.graph_dim, dd)
        self.decoder_time = nn.Parameter(torch.zeros(1, cfg.num_frames, 1, dd))
        self.decoder_blocks = nn.ModuleList([
            GraphSTB(dd, dh, cfg.num_landmarks, cfg.max_hop, cfg.mlp_ratio)
            for _ in range(cfg.decoder_depth)
        ])
        self.decoder_norm = nn.LayerNorm(dd)
        self.decoder_pred = nn.Linear(dd, 3)

        nn.init.trunc_normal_(self.graph_time_embed, std=0.02)
        nn.init.trunc_normal_(self.hyper_token, std=0.02)
        nn.init.trunc_normal_(self.decoder_time, std=0.02)

    def random_mask(self, B, T, N, device, ratio):
        """Vectorized (see VisionMAE.random_mask)."""
        n_mask = int(N * ratio)
        noise = torch.rand(B, T, N, device=device)
        thresh = noise.kthvalue(n_mask, dim=-1, keepdim=True).values
        return noise <= thresh

    def forward(self, landmarks_norm, ratio):
        B, T, N, _ = landmarks_norm.shape
        g = self.node_embed(landmarks_norm) + self.graph_time_embed
        mask = self.random_mask(B, T, N, landmarks_norm.device, ratio)
        g = self.mask_token(g, mask)
        hnode = self.hyper_token.expand(B, T, -1, -1)
        g = torch.cat([hnode, g], dim=2)  # (B,T,1+N,D), hyper-node always visible

        for blk in self.graph_blocks:
            g, _ = blk(g)
        g = self.enc_norm(g)

        d = self.decoder_embed(g) + torch.cat(
            [torch.zeros_like(self.decoder_time.expand(B, T, 1, -1)),
             self.decoder_time.expand(B, T, N, -1)], dim=2
        )
        for blk in self.decoder_blocks:
            d, _ = blk(d)
        d = self.decoder_norm(d)
        pred = self.decoder_pred(d)[:, :, 1:, :]  # drop hyper-node slot -> (B,T,N,3)

        loss_map = ((pred - landmarks_norm) ** 2).mean(dim=-1)
        loss = (loss_map * mask.float()).sum() / mask.float().sum().clamp(min=1)
        return loss


# %% CELL 15 -- training utilities: AMP state, warmup+cosine LR, NaN-safe step
class AmpState:
    """Bundles autocast dtype + (only-if-needed) GradScaler, and centralizes
    the non-finite-loss guard (fix E) so every training loop shares one
    code path instead of duplicating the skip-on-NaN logic three times."""

    def __init__(self, cfg: CFG):
        self.enabled = cfg.amp and cfg.device == "cuda"
        self.dtype = resolve_amp_dtype(cfg)
        self.use_scaler = self.enabled and self.dtype == torch.float16
        self.scaler = torch.amp.GradScaler("cuda", enabled=self.use_scaler)
        if self.enabled:
            print(f"[amp] enabled, dtype={self.dtype}, using GradScaler={self.use_scaler}")

    def autocast(self):
        return torch.amp.autocast("cuda", dtype=self.dtype, enabled=self.enabled)

    def backward_and_step(self, loss, model, optimizer, grad_clip) -> bool:
        """Returns True iff the optimizer actually stepped."""
        if not torch.isfinite(loss):
            optimizer.zero_grad(set_to_none=True)
            return False
        if self.use_scaler:
            self.scaler.scale(loss).backward()
            self.scaler.unscale_(optimizer)
            torch.nn.utils.clip_grad_norm_(model.parameters(), grad_clip)
            prev_scale = self.scaler.get_scale()
            self.scaler.step(optimizer)
            self.scaler.update()
            skipped = self.scaler.get_scale() < prev_scale  # GradScaler's own inf/nan-grad guard
            optimizer.zero_grad(set_to_none=True)
            return not skipped
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), grad_clip)
        optimizer.step()
        optimizer.zero_grad(set_to_none=True)
        return True


def build_warmup_cosine_scheduler(optimizer, total_steps, warmup_ratio=0.05, min_lr_ratio=0.0):
    """Fix D: the paper specifies warmup + cosine decay; the original draft
    used plain CosineAnnealingLR from step 0 (full LR immediately), a classic
    cause of early divergence for a from-scratch 24-layer transformer."""
    warmup_steps = max(1, int(total_steps * warmup_ratio))

    def lr_lambda(step):
        if step < warmup_steps:
            return (step + 1) / warmup_steps
        progress = min(1.0, (step - warmup_steps) / max(1, total_steps - warmup_steps))
        cosine = 0.5 * (1 + math.cos(math.pi * progress))
        return min_lr_ratio + (1 - min_lr_ratio) * cosine

    return torch.optim.lr_scheduler.LambdaLR(optimizer, lr_lambda)


def _check_nan_streak(nan_streak, max_nan_skips, where):
    if nan_streak > max_nan_skips:
        raise RuntimeError(
            f"{where} diverged: {nan_streak} consecutive non-finite losses. "
            "Try lowering the relevant LR in CFG, or CFG.guide_bias_clip."
        )


# %% CELL 16 -- pretraining loops (Stage 1, run independently per stream)
def pretrain_vision(cfg: CFG, dataset: Dataset):
    model = VisionMAE(cfg).to(cfg.device)
    loader = DataLoader(dataset, batch_size=cfg.pretrain_batch_size, shuffle=True,
                         num_workers=cfg.num_workers, drop_last=True, pin_memory=True)
    opt = torch.optim.Adam(model.parameters(), lr=cfg.pretrain_lr)
    sched = build_warmup_cosine_scheduler(opt, cfg.pretrain_epochs_vis * len(loader), cfg.warmup_ratio)
    amp_state = AmpState(cfg)

    model.train()
    nan_streak = 0
    for epoch in range(cfg.pretrain_epochs_vis):
        pbar = tqdm(loader, desc=f"[pretrain-vision] epoch {epoch+1}/{cfg.pretrain_epochs_vis}")
        for batch in pbar:
            frames = batch["frames"].to(cfg.device, non_blocking=True)
            with amp_state.autocast():
                loss = model(frames, cfg.pretrain_mask_ratio)
            stepped = amp_state.backward_and_step(loss, model, opt, cfg.grad_clip)
            if stepped:
                sched.step()
                nan_streak = 0
                pbar.set_postfix(loss=loss.item())
            else:
                nan_streak += 1
                pbar.set_postfix(status=f"skipped non-finite (#{nan_streak})")
                _check_nan_streak(nan_streak, cfg.max_nan_skips, "pretrain_vision")
    torch.save(model.state_dict(), os.path.join(cfg.ckpt_dir, "vision_mae.pt"))
    return model


def pretrain_graph(cfg: CFG, dataset: Dataset):
    model = GraphMAE(cfg).to(cfg.device)
    loader = DataLoader(dataset, batch_size=cfg.pretrain_batch_size, shuffle=True,
                         num_workers=cfg.num_workers, drop_last=True, pin_memory=True)
    opt = torch.optim.Adam(model.parameters(), lr=cfg.pretrain_lr)
    sched = build_warmup_cosine_scheduler(opt, cfg.pretrain_epochs_graph * len(loader), cfg.warmup_ratio)
    amp_state = AmpState(cfg)

    model.train()
    nan_streak = 0
    for epoch in range(cfg.pretrain_epochs_graph):
        pbar = tqdm(loader, desc=f"[pretrain-graph] epoch {epoch+1}/{cfg.pretrain_epochs_graph}")
        for batch in pbar:
            lmk = batch["landmarks_norm"].to(cfg.device, non_blocking=True)
            with amp_state.autocast():
                loss = model(lmk, cfg.pretrain_mask_ratio)
            stepped = amp_state.backward_and_step(loss, model, opt, cfg.grad_clip)
            if stepped:
                sched.step()
                nan_streak = 0
                pbar.set_postfix(loss=loss.item())
            else:
                nan_streak += 1
                pbar.set_postfix(status=f"skipped non-finite (#{nan_streak})")
                _check_nan_streak(nan_streak, cfg.max_nan_skips, "pretrain_graph")
    torch.save(model.state_dict(), os.path.join(cfg.ckpt_dir, "graph_mae.pt"))
    return model


def load_pretrained_vision(model: G2V2former, vis_mae: VisionMAE):
    model.patch_embed.load_state_dict(vis_mae.patch_embed.state_dict())
    model.vis_pos_embed.data.copy_(vis_mae.vis_pos_embed.data)
    model.vis_time_embed.data.copy_(vis_mae.vis_time_embed.data)
    for fb, pb in zip(model.vis_blocks, vis_mae.vis_blocks):
        fb.load_state_dict(pb.state_dict())


def load_pretrained_graph(model: G2V2former, graph_mae: GraphMAE):
    model.node_embed.load_state_dict(graph_mae.node_embed.state_dict())
    model.graph_time_embed.data.copy_(graph_mae.graph_time_embed.data)
    model.hyper_node.data.copy_(graph_mae.hyper_token.data)
    for fb, pb in zip(model.graph_blocks, graph_mae.graph_blocks):
        fb.load_state_dict(pb.state_dict())


# %% CELL 17 -- evaluation metrics
def compute_hter_auc(labels: np.ndarray, scores: np.ndarray):
    """labels: 1=live,0=spoof. scores: higher = more 'live'."""
    if roc_curve is None:
        raise ImportError("scikit-learn not importable -- check the offline package bundle")
    fpr, tpr, thr = roc_curve(labels, scores)
    fnr = 1 - tpr
    idx = np.nanargmin(np.abs(fnr - fpr))
    hter = (fpr[idx] + fnr[idx]) / 2 * 100
    auc_val = sk_auc(fpr, tpr) * 100
    return hter, auc_val


@torch.no_grad()
def evaluate(model, loader, cfg: CFG):
    model.eval()
    device = cfg.device
    amp_enabled = cfg.amp and device == "cuda"
    amp_dtype = resolve_amp_dtype(cfg)
    all_labels, all_scores = [], []
    for batch in loader:
        frames = batch["frames"].to(device)
        lmk_norm = batch["landmarks_norm"].to(device)
        lmk_px = batch["landmarks_px"].to(device)
        labels = batch["label"].to(device)
        with torch.amp.autocast("cuda", dtype=amp_dtype, enabled=amp_enabled):
            logits, *_ = model(frames, lmk_norm, lmk_px)
        probs = F.softmax(logits.float(), dim=-1)[:, 1]  # P(live)
        all_labels.append(labels.cpu().numpy())
        all_scores.append(probs.cpu().numpy())
    labels = np.concatenate(all_labels)
    scores = np.concatenate(all_scores)
    hter, auc_val = compute_hter_auc(labels, scores)
    model.train()
    return hter, auc_val


# %% CELL 18 -- fine-tuning loop (Stage 2), with per-epoch Celeb-DF-v2 cross-dataset eval
def finetune(cfg: CFG, train_ds: Dataset, test_ds: Dataset,
             vis_mae: VisionMAE = None, graph_mae: GraphMAE = None,
             celebdf_loader: DataLoader = None):
    model = G2V2former(cfg).to(cfg.device)
    if vis_mae is not None:
        load_pretrained_vision(model, vis_mae)
    if graph_mae is not None:
        load_pretrained_graph(model, graph_mae)

    train_loader = DataLoader(train_ds, batch_size=cfg.finetune_batch_size, shuffle=True,
                               num_workers=cfg.num_workers, drop_last=True, pin_memory=True)
    test_loader = DataLoader(test_ds, batch_size=cfg.finetune_batch_size, shuffle=False,
                              num_workers=cfg.num_workers, pin_memory=True)

    opt = torch.optim.Adam(model.parameters(), lr=cfg.finetune_lr)
    sched = build_warmup_cosine_scheduler(opt, cfg.finetune_epochs * len(train_loader), cfg.warmup_ratio)
    amp_state = AmpState(cfg)
    ce = nn.CrossEntropyLoss()

    best_hter, best_hter_celebdf = 100.0, 100.0
    nan_streak = 0
    for epoch in range(cfg.finetune_epochs):
        model.train()
        pbar = tqdm(train_loader, desc=f"[finetune] epoch {epoch+1}/{cfg.finetune_epochs}")
        for batch in pbar:
            frames = batch["frames"].to(cfg.device, non_blocking=True)
            lmk_norm = batch["landmarks_norm"].to(cfg.device, non_blocking=True)
            lmk_px = batch["landmarks_px"].to(cfg.device, non_blocking=True)
            labels = batch["label"].to(cfg.device, non_blocking=True)

            with amp_state.autocast():
                logits, _, _, vis_cls_seq = model(frames, lmk_norm, lmk_px)
                loss_ce = ce(logits, labels)
                loss_pcl = photometric_consistency_loss(vis_cls_seq, labels, tau=cfg.pcl_tau)
                loss = loss_ce + cfg.pcl_weight * loss_pcl

            stepped = amp_state.backward_and_step(loss, model, opt, cfg.grad_clip)
            if stepped:
                sched.step()
                nan_streak = 0
                pbar.set_postfix(ce=loss_ce.item(), pcl=loss_pcl.item())
            else:
                nan_streak += 1
                pbar.set_postfix(status=f"skipped non-finite (#{nan_streak})")
                _check_nan_streak(nan_streak, cfg.max_nan_skips, "finetune")

        hter, auc_val = evaluate(model, test_loader, cfg)
        print(f"epoch {epoch+1}: [LCC-FASD in-domain]        HTER={hter:.2f}%  AUC={auc_val:.2f}%")
        if hter < best_hter:
            best_hter = hter
            torch.save(model.state_dict(), os.path.join(cfg.ckpt_dir, "g2v2former_best.pt"))
            print(f"  -> new best in-domain checkpoint saved (HTER={hter:.2f}%)")

        if celebdf_loader is not None:
            hter_c, auc_c = evaluate(model, celebdf_loader, cfg)
            print(f"epoch {epoch+1}: [Celeb-DF-v2 cross-dataset]  HTER={hter_c:.2f}%  AUC={auc_c:.2f}%")
            if hter_c < best_hter_celebdf:
                best_hter_celebdf = hter_c
                torch.save(model.state_dict(), os.path.join(cfg.ckpt_dir, "g2v2former_best_celebdf.pt"))
                print(f"  -> new best Celeb-DF-v2 checkpoint saved (HTER={hter_c:.2f}%)")

    return model, best_hter, best_hter_celebdf


# %% CELL 19 -- main
def main():
    cfg = CFG()
    if cfg.device == "cuda":
        torch.backends.cudnn.benchmark = True

    lcc_index = index_lcc_fasd()
    lmk_cache = build_landmark_cache(lcc_index, cfg.landmark_cache_path, device=cfg.device,
                                      input_path=cfg.landmark_cache_input_path)

    train_ds = FASVideoDataset(lcc_index.get("train", []), cfg.img_size, cfg.num_frames,
                                augment=True, lmk_cache=lmk_cache)
    test_ds = FASVideoDataset(lcc_index.get("test", []), cfg.img_size, cfg.num_frames,
                               augment=False, lmk_cache=lmk_cache)
    print(f"train videos: {len(train_ds)}  test videos: {len(test_ds)}")

    celebdf_loader = None
    if cfg.eval_celebdf_every_epoch:
        celebdf_loader = build_celebdf_eval_loader(cfg)
        print(f"Celeb-DF-v2 eval videos: {len(celebdf_loader.dataset)}")

    vis_mae, graph_mae = None, None
    if cfg.do_pretrain:
        # Stage 1: independent MAE pretraining per stream (unlabeled -- uses train split only)
        vis_mae = pretrain_vision(cfg, train_ds)
        graph_mae = pretrain_graph(cfg, train_ds)

    # Stage 2: fine-tune the fused two-stream model, evaluating on both
    # datasets after every epoch.
    model, best_hter, best_hter_celebdf = finetune(
        cfg, train_ds, test_ds, vis_mae, graph_mae, celebdf_loader
    )
    print(f"Best in-domain (LCC-FASD) HTER: {best_hter:.2f}%")
    if celebdf_loader is not None:
        print(f"Best cross-dataset (Celeb-DF-v2) HTER: {best_hter_celebdf:.2f}%")
    return cfg, model, test_ds


if __name__ == "__main__":
    cfg, model, test_ds = main()


# %% CELL 20 -- visualize results (run after main(), reuses `model`/`cfg`/`test_ds` from CELL 19)
import matplotlib.pyplot as plt


def load_best_model(cfg: CFG, ckpt_name="g2v2former_best.pt") -> G2V2former:
    model = G2V2former(cfg).to(cfg.device)
    ckpt_path = os.path.join(cfg.ckpt_dir, ckpt_name)
    model.load_state_dict(torch.load(ckpt_path, map_location=cfg.device, weights_only=True))
    model.eval()
    return model


@torch.no_grad()
def collect_predictions(model, loader, cfg: CFG, max_samples=200):
    """Returns per-video predictions plus a few raw frames+landmarks for display."""
    device = cfg.device
    amp_enabled = cfg.amp and device == "cuda"
    amp_dtype = resolve_amp_dtype(cfg)
    all_labels, all_scores, viz_samples = [], [], []
    for batch in loader:
        frames = batch["frames"].to(device)
        lmk_norm = batch["landmarks_norm"].to(device)
        lmk_px = batch["landmarks_px"].to(device)
        labels = batch["label"].to(device)

        with torch.amp.autocast("cuda", dtype=amp_dtype, enabled=amp_enabled):
            logits, *_ = model(frames, lmk_norm, lmk_px)
        probs = F.softmax(logits.float(), dim=-1)[:, 1]  # P(live)
        preds = (probs > 0.5).long()

        all_labels.append(labels.cpu().numpy())
        all_scores.append(probs.cpu().numpy())

        if len(viz_samples) < max_samples:
            for i in range(frames.shape[0]):
                viz_samples.append({
                    "frame0": frames[i, 0].cpu(),          # first frame of clip, (3,H,W), normalized [-1,1]
                    "lmk0_px": lmk_px[i, 0].cpu().numpy(),  # (68,2)
                    "label": labels[i].item(),
                    "pred": preds[i].item(),
                    "score": probs[i].item(),
                })
                if len(viz_samples) >= max_samples:
                    break

    labels = np.concatenate(all_labels)
    scores = np.concatenate(all_scores)
    return labels, scores, viz_samples


def denorm_img(t):
    """(3,H,W) in [-1,1] -> (H,W,3) uint8 RGB for matplotlib."""
    img = (t.permute(1, 2, 0).numpy() * 0.5 + 0.5).clip(0, 1)
    return (img * 255).astype(np.uint8)


def plot_sample_grid(viz_samples, n=12, save_path=None):
    """Grid of frames with landmark overlay; green border = correct, red = wrong."""
    n = min(n, len(viz_samples))
    cols = 4
    rows = math.ceil(n / cols)
    fig, axes = plt.subplots(rows, cols, figsize=(4 * cols, 4 * rows))
    axes = np.array(axes).reshape(-1)

    for i in range(rows * cols):
        ax = axes[i]
        ax.axis("off")
        if i >= n:
            continue
        s = viz_samples[i]
        img = denorm_img(s["frame0"])
        ax.imshow(img)
        ax.scatter(s["lmk0_px"][:, 0], s["lmk0_px"][:, 1], s=4, c="cyan", alpha=0.7)

        correct = s["label"] == s["pred"]
        label_str = "live" if s["label"] == 1 else "spoof"
        pred_str = "live" if s["pred"] == 1 else "spoof"
        color = "limegreen" if correct else "red"
        for spine in ax.spines.values():
            spine.set_visible(True)
            spine.set_color(color)
            spine.set_linewidth(4)
        ax.set_title(f"true={label_str} pred={pred_str} (p_live={s['score']:.2f})",
                     fontsize=9, color=color)

    fig.suptitle("Sample predictions (first frame of each clip, cyan = detected landmarks)", fontsize=13)
    fig.tight_layout()
    if save_path:
        fig.savefig(save_path, dpi=150, bbox_inches="tight")
    plt.show()


def plot_roc_and_scores(labels, scores, save_path=None):
    fpr, tpr, thr = roc_curve(labels, scores)
    fnr = 1 - tpr
    idx = np.nanargmin(np.abs(fnr - fpr))
    eer_thr = thr[idx]
    hter, auc_val = compute_hter_auc(labels, scores)

    fig, axes = plt.subplots(1, 2, figsize=(12, 5))

    axes[0].plot(fpr, tpr, label=f"AUC = {auc_val:.2f}%")
    axes[0].plot([0, 1], [0, 1], "k--", alpha=0.3)
    axes[0].scatter(fpr[idx], tpr[idx], c="red", zorder=5, label=f"EER point (HTER={hter:.2f}%)")
    axes[0].set_xlabel("False Positive Rate")
    axes[0].set_ylabel("True Positive Rate")
    axes[0].set_title("ROC curve")
    axes[0].legend()

    live_scores = scores[labels == 1]
    spoof_scores = scores[labels == 0]
    axes[1].hist(spoof_scores, bins=30, alpha=0.6, label="spoof (y=0)", color="tomato")
    axes[1].hist(live_scores, bins=30, alpha=0.6, label="live (y=1)", color="seagreen")
    axes[1].axvline(eer_thr, color="black", linestyle="--", label=f"EER threshold={eer_thr:.2f}")
    axes[1].set_xlabel("P(live) score")
    axes[1].set_ylabel("count")
    axes[1].set_title("Score distribution by class")
    axes[1].legend()

    fig.tight_layout()
    if save_path:
        fig.savefig(save_path, dpi=150, bbox_inches="tight")
    plt.show()
    return hter, auc_val, eer_thr


def plot_confusion_matrix(labels, scores, threshold=0.5, save_path=None):
    preds = (scores > threshold).astype(int)
    classes = ["spoof", "live"]
    cm = np.zeros((2, 2), dtype=int)
    for t, p in zip(labels, preds):
        cm[t, p] += 1

    fig, ax = plt.subplots(figsize=(4.5, 4))
    im = ax.imshow(cm, cmap="Blues")
    ax.set_xticks([0, 1]); ax.set_xticklabels(classes)
    ax.set_yticks([0, 1]); ax.set_yticklabels(classes)
    ax.set_xlabel("predicted"); ax.set_ylabel("true")
    ax.set_title(f"Confusion matrix (threshold={threshold})")
    for i in range(2):
        for j in range(2):
            ax.text(j, i, str(cm[i, j]), ha="center", va="center",
                     color="white" if cm[i, j] > cm.max() / 2 else "black", fontsize=14)
    fig.colorbar(im, fraction=0.046, pad=0.04)
    fig.tight_layout()
    if save_path:
        fig.savefig(save_path, dpi=150, bbox_inches="tight")
    plt.show()


def visualize_results(cfg: CFG, model: G2V2former = None, ds: Dataset = None,
                       ckpt_name="g2v2former_best.pt", tag="LCC-FASD"):
    if model is None:
        model = load_best_model(cfg, ckpt_name)
    if ds is None:
        lcc_index = index_lcc_fasd()
        lmk_cache = LandmarkCache(cfg.landmark_cache_path, input_path=cfg.landmark_cache_input_path)
        ds = FASVideoDataset(lcc_index.get("test", []), cfg.img_size, cfg.num_frames,
                              augment=False, lmk_cache=lmk_cache)

    loader = DataLoader(ds, batch_size=cfg.finetune_batch_size, shuffle=True,  # shuffled just for a varied sample grid
                         num_workers=cfg.num_workers)

    labels, scores, viz_samples = collect_predictions(model, loader, cfg)

    plot_sample_grid(viz_samples, n=12,
                      save_path=os.path.join(cfg.ckpt_dir, f"sample_predictions_{tag}.png"))
    hter, auc_val, eer_thr = plot_roc_and_scores(
        labels, scores, save_path=os.path.join(cfg.ckpt_dir, f"roc_scores_{tag}.png"))
    plot_confusion_matrix(labels, scores, threshold=eer_thr,
                           save_path=os.path.join(cfg.ckpt_dir, f"confusion_matrix_{tag}.png"))

    print(f"[{tag}] Final: HTER={hter:.2f}%  AUC={auc_val:.2f}%  (threshold @ EER = {eer_thr:.3f})")


# In-domain visualization (reuses `model`/`test_ds`/`cfg` from CELL 19 if still in scope, else reloads)
visualize_results(
    cfg=cfg if "cfg" in dir() else CFG(),
    model=model if "model" in dir() else None,
    ds=test_ds if "test_ds" in dir() else None,
    ckpt_name="g2v2former_best.pt", tag="LCC-FASD",
)

# Cross-dataset (Celeb-DF-v2, deepfake) visualization, using its own best checkpoint
_celebdf_ckpt = os.path.join((cfg if "cfg" in dir() else CFG()).ckpt_dir, "g2v2former_best_celebdf.pt")
if os.path.exists(_celebdf_ckpt):
    _cfg = cfg if "cfg" in dir() else CFG()
    _celebdf_index = build_celebdf_index_cached(_cfg)
    _celebdf_ds = CelebDFVideoDataset(_celebdf_index, img_size=_cfg.img_size, num_frames=_cfg.num_frames)
    visualize_results(cfg=_cfg, model=None, ds=_celebdf_ds,
                       ckpt_name="g2v2former_best_celebdf.pt", tag="CelebDF-v2")


# %% CELL 21 -- cleanup
# !rm -rf /kaggle/working/hub /kaggle/working/offline_site_packages
# !rm -rf /kaggle/working/cache   # landmark cache + celebdf index -- small, cheap to regenerate
# Nothing else lives under /kaggle/working by design (fix J): LCC-FASD and
# Celeb-DF-v2 frames are always read directly from /kaggle/input, never
# copied, so there's no multi-GB "dataset" folder to clean up here. Leave
# /kaggle/working/ckpts alone until you've saved off the checkpoints/plots
# you want.
```

```python
# %% CELL 22 -- index the Asian-Fakes dataset (single-frame stills, not clips)
# Layout expected:
#   <root>/msib_asian/real/*.png|jpg          (live)
#   <root>/msib_asian/sbi/*.png|jpg           (spoof / deepfake)
#   <root>/msib_non_asian/real/*.png|jpg      (live)
#   <root>/msib_non_asian/sbi/*.png|jpg       (spoof / deepfake)
#
# These are individual frames, not video clips -- but FASVideoDataset
# already has a "clip shorter than num_frames" fallback (repeat the last
# available frame) baked into `_sample_frame_indices`, so treating every
# still image as a 1-frame "clip" and reusing FASVideoDataset UNCHANGED
# gives exactly the right behavior: the same still gets replicated across
# all T temporal slots (KTA/temporal attention degrades gracefully to
# "no motion", which is the correct thing to feed a video model a static
# image). No new Dataset class needed.

class LandmarkCache:
    def __init__(self, cache_path, input_path=None):
        """Loads from `cache_path` (writable, /kaggle/working) if it already
        exists there (e.g. a resumed session); otherwise seeds from
        `input_path` if given -- a cache you published as a Kaggle Dataset
        on an earlier run, mounted read-only under /kaggle/input. Either
        way, `save()` always writes to `cache_path`; /kaggle/input can't be
        written to mid-session."""
        self.cache_path = cache_path
        if os.path.exists(cache_path):
            self.data = torch.load(cache_path, weights_only=False)
            print(f"[lmk-cache] loaded {len(self.data)} cached landmarks from {cache_path}")
        elif input_path and os.path.exists(input_path):
            self.data = torch.load(input_path, weights_only=False)
            print(f"[lmk-cache] seeded {len(self.data)} landmarks from published input cache {input_path}")
        else:
            self.data = {}

    def __contains__(self, key):
        return key in self.data

    def __len__(self):
        return len(self.data)

    def get(self, key):
        return self.data.get(key)

    def set(self, key, lmk):
        self.data[key] = lmk

    def save(self):
        os.makedirs(os.path.dirname(self.cache_path), exist_ok=True)
        torch.save(self.data, self.cache_path)

ASIAN_FAKES_ROOT = "/kaggle/input/notebooks/raihanzahin/prep-sbi-extra/sbi_faces_extra/images"  # EDIT to match your attached dataset path
ASIAN_LANDMARK_CACHE_PATH = "/kaggle/working/cache/asian_landmarks.pt"

ASIAN_SUBGROUPS = ["msib_asian", "msib_non_asian"]
ASIAN_LABEL_MAP = {"real": "live", "sbi": "attack"}          # matches LCC_LABEL_MAP / CelebDF convention: 1=live,0=spoof
ASIAN_IMG_EXTS = (".png", ".jpg", ".jpeg", ".bmp", ".webp")


def index_asian_fakes(root=ASIAN_FAKES_ROOT, subgroups=ASIAN_SUBGROUPS, label_map=ASIAN_LABEL_MAP):
    """Returns {"all": [...], "msib_asian": [...], "msib_non_asian": [...]},
    each a flat list of (sample_id, [single abs frame path], label_str) in
    the exact tuple shape FASVideoDataset/index_lcc_fasd already expect.
    Nothing is copied -- frames stay wherever Kaggle mounted them."""
    out = {"all": []}
    for sub in subgroups:
        sub_path = os.path.join(root, sub)
        out[sub] = []
        if not os.path.isdir(sub_path):
            print(f"[asian-fakes] WARNING: {sub_path} not found, skipping.")
            continue
        for src_label, dst_label in label_map.items():
            src_path = os.path.join(sub_path, src_label)
            if not os.path.isdir(src_path):
                print(f"[asian-fakes] WARNING: {src_path} not found, skipping.")
                continue
            n = 0
            for fname in sorted(os.listdir(src_path)):
                if fname.lower().endswith(ASIAN_IMG_EXTS):
                    fpath = os.path.join(src_path, fname)
                    sample_id = f"asian_{sub}_{dst_label}_{os.path.splitext(fname)[0]}"
                    entry = (sample_id, [fpath], dst_label)
                    out[sub].append(entry)
                    out["all"].append(entry)
                    n += 1
            print(f"[asian-fakes] {sub}/{dst_label}: {n} images indexed (0 bytes copied)")
    print(f"[asian-fakes] total: {len(out['all'])} images "
          f"({len(out.get('msib_asian', []))} asian, {len(out.get('msib_non_asian', []))} non-asian)")
    return out


ASIAN_INDEX = index_asian_fakes()

# Reuses build_landmark_cache from CELL 4 unchanged -- it just wants a
# {split_name: [(id, paths, label), ...]} dict, which is exactly what
# index_asian_fakes() returns. Cached to its own file so it never collides
# with the LCC-FASD landmark cache.
asian_lmk_cache = build_landmark_cache(
    {"all": ASIAN_INDEX["all"]}, ASIAN_LANDMARK_CACHE_PATH, device=CFG.device,
)

asian_test_ds_all = FASVideoDataset(ASIAN_INDEX["all"], CFG.img_size, CFG.num_frames,
                                     augment=False, lmk_cache=asian_lmk_cache)
asian_test_ds_asian = FASVideoDataset(ASIAN_INDEX["msib_asian"], CFG.img_size, CFG.num_frames,
                                       augment=False, lmk_cache=asian_lmk_cache)
asian_test_ds_non_asian = FASVideoDataset(ASIAN_INDEX["msib_non_asian"], CFG.img_size, CFG.num_frames,
                                           augment=False, lmk_cache=asian_lmk_cache)

print(f"asian-fakes eval images: all={len(asian_test_ds_all)}  "
      f"asian={len(asian_test_ds_asian)}  non_asian={len(asian_test_ds_non_asian)}")
```

```python
# %% CELL 23 -- Celeb-DF-v2 TRAIN split (videos NOT in the official test protocol)
# CELL 6 only ever indexed the official `List_of_testing_videos.txt` protocol
# (for cross-dataset EVAL). To finetune on Celeb-DF-v2 we need a disjoint
# TRAIN split. Reuses every helper CELL 6 already defines
# (_find_celebdf_root, _crop_box_and_landmarks, _stratified_subset,
# load_celebdf_protocol) -- nothing in CELL 6 is modified, this only adds a
# sibling code path for "the other videos".

CELEBDF_TRAIN_INDEX_PATH = "/kaggle/working/cache/celebdf_train_index.pt"
CELEBDF_TRAIN_VIDEO_LIMIT = 300   # stratified cap so Stage-2 finetuning stays affordable; raise if time/VRAM allow


def load_celebdf_all_items(celebdf_root):
    items = []
    for folder, label in [("Celeb-real", "live"), ("YouTube-real", "live"), ("Celeb-synthesis", "attack")]:
        folder_path = os.path.join(celebdf_root, folder)
        if not os.path.isdir(folder_path):
            continue
        for fname in sorted(os.listdir(folder_path)):
            if fname.lower().endswith(".mp4"):
                items.append((f"{folder}/{fname}", label))
    return items


def load_celebdf_train_items(celebdf_root):
    """All videos EXCLUDING the official test protocol list. The test
    protocol videos are left completely untouched so the Celeb-DF-v2
    cross-dataset numbers from CELL 18/20 stay comparable before/after this
    experiment."""
    test_items = load_celebdf_protocol(celebdf_root)
    test_paths = {rel for rel, _ in test_items}
    all_items = load_celebdf_all_items(celebdf_root)
    train_items = [(rel, label) for rel, label in all_items if rel not in test_paths]
    print(f"[celebdf] train split (all videos minus official test protocol): "
          f"{len(train_items)} videos (test protocol has {len(test_items)})")
    return train_items


def build_celebdf_index_from_items(items, celebdf_root, fa_model, num_frames_to_cache=16,
                                    img_size=224, crop_margin=0.4, limit=None,
                                    desc="celebdf: indexing (metadata only, no frames written)"):
    """Factored out of CELL 6's build_celebdf_index so the same metadata-only
    (crop box + landmarks, never pixels) indexing logic can run over an
    arbitrary item list instead of always re-deriving the official test
    protocol from disk."""
    if limit:
        items = _stratified_subset(items, limit)
    index = []
    for rel_path, label in tqdm(items, desc=desc):
        video_path = os.path.join(celebdf_root, rel_path)
        if not os.path.isfile(video_path):
            continue
        cap = cv2.VideoCapture(video_path)
        total = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
        if total <= 0:
            cap.release()
            continue
        target_set = set(np.linspace(0, total - 1, min(num_frames_to_cache, total)).round().astype(int).tolist())

        frame_meta, i = [], 0
        while len(frame_meta) < len(target_set):
            ok = cap.grab()
            if not ok:
                break
            if i in target_set:
                ok, frame = cap.retrieve()
                if ok:
                    h, w = frame.shape[:2]
                    frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
                    try:
                        preds = fa_model.get_landmarks(frame_rgb)
                    except Exception:
                        preds = None
                    box, lmk = _crop_box_and_landmarks(preds, h, w, crop_margin, img_size)
                    frame_meta.append({"frame_idx": i, "box": box, "lmk_crop": lmk})
            i += 1
        cap.release()
        if frame_meta:
            index.append({
                "video_path": video_path,
                "label": 1 if label == "live" else 0,
                "frames": frame_meta,
            })
    print(f"[celebdf] indexed {len(index)}/{len(items)} videos (metadata-only, no frame files written)")
    return index


def build_celebdf_train_index_cached(cfg: "CFG", cache_path=CELEBDF_TRAIN_INDEX_PATH,
                                      video_limit=CELEBDF_TRAIN_VIDEO_LIMIT):
    """Analogous to CELL 6's build_celebdf_index_cached, but for the TRAIN
    split. Separate cache file -- never collides with the eval index."""
    if os.path.exists(cache_path):
        index = torch.load(cache_path, weights_only=False)  # our own cache, not a state_dict
        print(f"[celebdf-train] loaded cached index: {len(index)} videos from {cache_path}")
        return index

    import face_alignment
    celebdf_root = _find_celebdf_root()
    train_items = load_celebdf_train_items(celebdf_root)
    fa_model = face_alignment.FaceAlignment(
        face_alignment.LandmarksType.THREE_D, flip_input=False, device=cfg.device
    )
    index = build_celebdf_index_from_items(
        train_items, celebdf_root, fa_model,
        num_frames_to_cache=cfg.celebdf_frames_per_video, img_size=cfg.img_size,
        crop_margin=cfg.celebdf_crop_margin, limit=video_limit,
        desc="celebdf-train: indexing (metadata only, no frames written)",
    )
    del fa_model
    if cfg.device == "cuda":
        torch.cuda.empty_cache()

    os.makedirs(os.path.dirname(cache_path), exist_ok=True)
    torch.save(index, cache_path)
    return index


_cfg_for_index = cfg if "cfg" in dir() else CFG()
celebdf_train_index = build_celebdf_train_index_cached(_cfg_for_index, CELEBDF_TRAIN_INDEX_PATH, CELEBDF_TRAIN_VIDEO_LIMIT)
celebdf_train_ds = CelebDFVideoDataset(celebdf_train_index, img_size=_cfg_for_index.img_size, num_frames=_cfg_for_index.num_frames)
print(f"Celeb-DF-v2 TRAIN videos (for Stage-2 finetuning): {len(celebdf_train_ds)}")
```

```python
# %% CELL 24 -- generic multi-dataset evaluation + "continue finetuning from
# an existing checkpoint on a new dataset" (needed for experiment 2: LCC-FASD
# -> finetune on Celeb-DF-v2). Both reuse evaluate()/G2V2former/AmpState/
# build_warmup_cosine_scheduler/photometric_consistency_loss/_check_nan_streak
# from CELLS 12-18 unchanged.

def run_full_eval(model, loaders: dict, cfg: "CFG", tag: str):
    """loaders: {display_name: DataLoader}. Every dataset class in this
    notebook (FASVideoDataset, CelebDFVideoDataset) yields the same batch
    keys, so evaluate() works unmodified across all of them."""
    print(f"\n===== Evaluating [{tag}] on all datasets =====")
    results = {}
    for name, loader in loaders.items():
        hter, auc_val = evaluate(model, loader, cfg)
        results[name] = {"HTER": hter, "AUC": auc_val}
        print(f"[{tag}] {name:24s} HTER={hter:6.2f}%  AUC={auc_val:6.2f}%")
    return results


def finetune_continue(cfg: "CFG", train_ds: Dataset, eval_loaders: dict, init_ckpt_path: str,
                       primary_key: str, epochs: int = None, lr: float = None, tag: str = "stage2"):
    """Loads a FULL previously fine-tuned G2V2former checkpoint (not just the
    MAE sub-modules, unlike CELL 18's `finetune`) and continues supervised
    training on `train_ds`, evaluating on every loader in `eval_loaders`
    after each epoch. Saves the checkpoint with the best HTER on
    `eval_loaders[primary_key]` to f'{cfg.ckpt_dir}/g2v2former_{tag}_best.pt'.
    Mirrors CELL 18's training loop (same losses, AMP/NaN-guard, warmup+cosine
    schedule) but generalized to an arbitrary dict of eval loaders instead of
    a single hardcoded Celeb-DF-v2 loader, and to loading full-model weights
    instead of MAE pretrain weights.
    """
    model = G2V2former(cfg).to(cfg.device)
    state = torch.load(init_ckpt_path, map_location=cfg.device, weights_only=False)
    model.load_state_dict(state)
    print(f"[finetune_continue:{tag}] loaded init weights from {init_ckpt_path}")

    epochs = epochs if epochs is not None else cfg.finetune_epochs
    lr = lr if lr is not None else cfg.finetune_lr

    train_loader = DataLoader(train_ds, batch_size=cfg.finetune_batch_size, shuffle=True,
                               num_workers=cfg.num_workers, drop_last=True, pin_memory=True)

    opt = torch.optim.Adam(model.parameters(), lr=lr)
    sched = build_warmup_cosine_scheduler(opt, epochs * len(train_loader), cfg.warmup_ratio)
    amp_state = AmpState(cfg)
    ce = nn.CrossEntropyLoss()

    best_primary = 100.0
    history = []
    nan_streak = 0
    for epoch in range(epochs):
        model.train()
        pbar = tqdm(train_loader, desc=f"[{tag}] epoch {epoch+1}/{epochs}")
        for batch in pbar:
            frames = batch["frames"].to(cfg.device, non_blocking=True)
            lmk_norm = batch["landmarks_norm"].to(cfg.device, non_blocking=True)
            lmk_px = batch["landmarks_px"].to(cfg.device, non_blocking=True)
            labels = batch["label"].to(cfg.device, non_blocking=True)

            with amp_state.autocast():
                logits, _, _, vis_cls_seq = model(frames, lmk_norm, lmk_px)
                loss_ce = ce(logits, labels)
                loss_pcl = photometric_consistency_loss(vis_cls_seq, labels, tau=cfg.pcl_tau)
                loss = loss_ce + cfg.pcl_weight * loss_pcl

            stepped = amp_state.backward_and_step(loss, model, opt, cfg.grad_clip)
            if stepped:
                sched.step()
                nan_streak = 0
                pbar.set_postfix(ce=loss_ce.item(), pcl=loss_pcl.item())
            else:
                nan_streak += 1
                pbar.set_postfix(status=f"skipped non-finite (#{nan_streak})")
                _check_nan_streak(nan_streak, cfg.max_nan_skips, tag)

        epoch_result = {"epoch": epoch + 1}
        for name, loader in eval_loaders.items():
            hter, auc_val = evaluate(model, loader, cfg)
            epoch_result[name] = {"HTER": hter, "AUC": auc_val}
            print(f"epoch {epoch+1}: [{name}] HTER={hter:.2f}%  AUC={auc_val:.2f}%")
        history.append(epoch_result)

        primary_hter = epoch_result[primary_key]["HTER"]
        if primary_hter < best_primary:
            best_primary = primary_hter
            ckpt_path = os.path.join(cfg.ckpt_dir, f"g2v2former_{tag}_best.pt")
            torch.save(model.state_dict(), ckpt_path)
            print(f"  -> new best [{tag}] checkpoint saved "
                  f"(primary metric = HTER on '{primary_key}' = {primary_hter:.2f}%)")

    return model, history
```

```python
# %% CELL 25 -- EXPERIMENT 1: train on LCC-FASD only, test on all 3 datasets
# Reuses the run from CELL 19 (`cfg`/`model`/`test_ds`) if it's still in
# scope in this kernel session; otherwise (re)runs main() to produce a
# LCC-FASD-trained checkpoint first.

if all(name in dir() for name in ("cfg", "model", "test_ds")):
    exp1_cfg, exp1_test_ds = cfg, test_ds
else:
    exp1_cfg, _, exp1_test_ds = main()

# Always evaluate off the saved BEST checkpoint (not necessarily the
# in-memory weights from the final epoch) -- same convention CELL 20 uses.
exp1_model = load_best_model(exp1_cfg, "g2v2former_best.pt")

# Celeb-DF-v2 eval loader: reuse if CELL 19 already built one, else build now.
if "celebdf_loader" in dir() and celebdf_loader is not None:
    exp1_celebdf_loader = celebdf_loader
else:
    exp1_celebdf_loader = build_celebdf_eval_loader(exp1_cfg)

EVAL_LOADERS = {
    "LCC-FASD (test)": DataLoader(exp1_test_ds, batch_size=exp1_cfg.finetune_batch_size, shuffle=False,
                                   num_workers=exp1_cfg.num_workers, pin_memory=True),
    "CelebDF-v2 (test)": exp1_celebdf_loader,
    "Asian-Fakes (all)": DataLoader(asian_test_ds_all, batch_size=exp1_cfg.finetune_batch_size, shuffle=False,
                                     num_workers=exp1_cfg.num_workers, pin_memory=True),
    "Asian-Fakes (msib_asian)": DataLoader(asian_test_ds_asian, batch_size=exp1_cfg.finetune_batch_size, shuffle=False,
                                            num_workers=exp1_cfg.num_workers, pin_memory=True),
    "Asian-Fakes (msib_non_asian)": DataLoader(asian_test_ds_non_asian, batch_size=exp1_cfg.finetune_batch_size,
                                                shuffle=False, num_workers=exp1_cfg.num_workers, pin_memory=True),
}
# EVAL_LOADERS is reused as-is by experiment 2 (CELL 26) and the
# visualization cell (CELL 27), so both experiments are scored on the exact
# same held-out samples.

ALL_RESULTS = {}
ALL_RESULTS["Exp1: train=LCC-FASD"] = run_full_eval(
    exp1_model, EVAL_LOADERS, exp1_cfg, tag="Exp1: train=LCC-FASD"
)
```

```python
# %% CELL 27 -- visualize & compare EXPERIMENT 1 vs EXPERIMENT 2 across all
# 3 datasets. Builds a results table + grouped bar charts (HTER, AUC), and
# (optionally) reuses CELL 20's ROC/confusion-matrix plotting for each
# (experiment, dataset) pair.
import pandas as pd


def results_to_dataframe(all_results: dict) -> pd.DataFrame:
    rows = []
    for exp_name, ds_results in all_results.items():
        for ds_name, metrics in ds_results.items():
            rows.append({"experiment": exp_name, "dataset": ds_name,
                         "HTER (%)": metrics["HTER"], "AUC (%)": metrics["AUC"]})
    return pd.DataFrame(rows)


results_df = results_to_dataframe(ALL_RESULTS)
pivot = results_df.pivot(index="dataset", columns="experiment", values=["HTER (%)", "AUC (%)"]).round(2)
print(pivot)

csv_path = os.path.join(exp1_cfg.ckpt_dir, "experiment_comparison.csv")
results_df.to_csv(csv_path, index=False)
print(f"\nSaved results table -> {csv_path}")


def plot_metric_comparison(df: pd.DataFrame, metric: str, save_path=None):
    datasets = df["dataset"].unique().tolist()
    experiments = df["experiment"].unique().tolist()
    x = np.arange(len(datasets))
    width = 0.8 / max(1, len(experiments))

    fig, ax = plt.subplots(figsize=(1.8 * len(datasets) + 3, 5))
    for i, exp in enumerate(experiments):
        vals = []
        for d in datasets:
            row = df[(df.dataset == d) & (df.experiment == exp)]
            vals.append(row[metric].values[0] if len(row) else np.nan)
        offset = i * width - width * (len(experiments) - 1) / 2
        bars = ax.bar(x + offset, vals, width, label=exp)
        ax.bar_label(bars, fmt="%.1f", fontsize=8, padding=2)

    ax.set_xticks(x)
    ax.set_xticklabels(datasets, rotation=20, ha="right")
    ax.set_ylabel(metric)
    ax.set_title(f"{metric} by dataset and training regime")
    ax.legend(fontsize=9)
    fig.tight_layout()
    if save_path:
        fig.savefig(save_path, dpi=150, bbox_inches="tight")
    plt.show()


plot_metric_comparison(results_df, "HTER (%)", save_path=os.path.join(exp1_cfg.ckpt_dir, "comparison_hter.png"))
plot_metric_comparison(results_df, "AUC (%)", save_path=os.path.join(exp1_cfg.ckpt_dir, "comparison_auc.png"))


# ---- optional: full ROC/score-distribution/confusion-matrix plots for each
# (experiment, dataset) pair, reusing CELL 20's plotting functions directly
# on a DataLoader instead of a freshly-wrapped Dataset.
def visualize_results_from_loader(cfg: "CFG", model: G2V2former, loader: DataLoader, tag: str):
    labels, scores, viz_samples = collect_predictions(model, loader, cfg)
    plot_sample_grid(viz_samples, n=12,
                      save_path=os.path.join(cfg.ckpt_dir, f"sample_predictions_{tag}.png"))
    hter, auc_val, eer_thr = plot_roc_and_scores(
        labels, scores, save_path=os.path.join(cfg.ckpt_dir, f"roc_scores_{tag}.png"))
    plot_confusion_matrix(labels, scores, threshold=eer_thr,
                           save_path=os.path.join(cfg.ckpt_dir, f"confusion_matrix_{tag}.png"))
    print(f"[{tag}] Final: HTER={hter:.2f}%  AUC={auc_val:.2f}%  (threshold @ EER = {eer_thr:.3f})")


# Set to True to also render the full per-dataset ROC/confusion-matrix grids
# (adds several plots per experiment x dataset combination).
RUN_DETAILED_PLOTS = False
if RUN_DETAILED_PLOTS:
    for exp_tag, exp_model in [("exp1_lcc_only", exp1_model), ("exp2_lcc_then_celebdf", exp2_best_model)]:
        for ds_name, loader in EVAL_LOADERS.items():
            safe_ds_name = ds_name.replace(" ", "_").replace("(", "").replace(")", "")
            visualize_results_from_loader(exp1_cfg, exp_model, loader, tag=f"{exp_tag}__{safe_ds_name}")
```
